## 1. String Methods for Data CleaningReal-world data is messy. String methods help clean it.

```python
# Demo: Essential String Cleaning Methods
# Real-world data is messy - these methods help standardize text

# 1. strip() - Remove leading and trailing whitespace
# Syntax: string.strip()
# Common issue: data copied from spreadsheets often has extra spaces
product_name = "  Suite A  "
clean_name = product_name.strip()  # Removes spaces from both ends
print(f"'{product_name}' -> '{clean_name}'")

# 2. upper() and lower() - Change case
# Syntax: string.upper() or string.lower()
# Useful for case-insensitive comparisons
region = "emea"
print(f"Upper: {region.upper()}")  # Returns "EMEA"

# 3. replace() - Substitute characters or substrings
# Syntax: string.replace(old, new)
# Replaces ALL occurrences of 'old' with 'new'
code = "SUITE-A-2024"
clean_code = code.replace("-", "_")  # Replace hyphens with underscores
print(f"Cleaned: {clean_code}")
```

### Exercise: Clean product names

```python
# Exercise: Clean Product Names
# Task: Standardize a list of inconsistently formatted product names
# Goal: All uppercase, spaces instead of hyphens/underscores, no extra whitespace

products = ["  suite a  ", "SUITE-B", "Suite_C  "]
cleaned = []

# Process each product name
for product in products:
    # Chain multiple string methods together (left to right):
    # 1. strip() - remove leading/trailing whitespace
    # 2. upper() - convert to uppercase
    # 3. replace("-", " ") - replace hyphens with spaces
    # 4. replace("_", " ") - replace underscores with spaces
    clean = product.strip().upper().replace("-", " ").replace("_", " ")
    cleaned.append(clean)

print(cleaned)  # Should output: ['SUITE A', 'SUITE B', 'SUITE C']
```

## 2. Checking String Properties

Methods that return True/False for checking strings.

```python
# Demo: Checking String Properties
# Methods that return True/False - useful for validation and filtering

# 1. startswith() and endswith() - Check beginning or end of string
# Syntax: string.startswith("prefix") or string.endswith("suffix")
product_code = "SUITE-A-2024"

print(f"Starts with 'SUITE': {product_code.startswith('SUITE')}")  # True
print(f"Ends with '2024': {product_code.endswith('2024')}")        # True
print(f"Ends with '2023': {product_code.endswith('2023')}")        # False

# 2. isdigit() - Check if string contains only digits
# Syntax: string.isdigit()
# Returns True only if ALL characters are digits (0-9)
order_id = "12345"
description = "Premium Suite"

print(f"\n'{order_id}' is numeric: {order_id.isdigit()}")       # True
print(f"'{description}' is numeric: {description.isdigit()}")   # False

# Trick: Check if text is alphabetic (remove spaces first)
print(f"'{description}' contains only letters and spaces: {description.replace(' ', '').isalpha()}")  # True

# 3. 'in' operator - Check if substring exists
# Syntax: "substring" in string
# Returns True if substring is found anywhere in string
email = "john.smith@acme.com"
print(f"\nEmail contains '@': {'@' in email}")      # True
print(f"Email contains 'acme': {'acme' in email}")  # True
```

## String Methods Quick Reference

| Method | Description | Example |
|--------|-------------|---------|
| `str.upper()` | Convert to uppercase | `"hello".upper()` → `"HELLO"` |
| `str.lower()` | Convert to lowercase | `"HELLO".lower()` → `"hello"` |
| `str.strip()` | Remove leading/trailing whitespace | `"  hi  ".strip()` → `"hi"` |
| `str.replace(old, new)` | Replace text | `"a-b".replace("-", "_")` → `"a_b"` |
| `str.split(sep)` | Split into list | `"a,b,c".split(",")` → `["a","b","c"]` |
| `sep.join(list)` | Join list into string | `",".join(["a","b"])` → `"a,b"` |
| `str.startswith(prefix)` | Check start | `"PROD-01".startswith("PROD")` → `True` |
| `str.endswith(suffix)` | Check end | `"file.csv".endswith(".csv")` → `True` |
| `str.isdigit()` | Check if all digits | `"123".isdigit()` → `True` |
| `"x" in str` | Check if contains | `"@" in "a@b.com"` → `True` |

```python
# Demo: Splitting and Joining Strings
# Essential for parsing structured text (CSV, paths, etc.)

# 1. split() - Break a string into a list
# Syntax: string.split("separator")
# Returns a list of substrings
csv_line = "Suite A,199,150"
fields = csv_line.split(",")  # Split on comma
print(f"Fields: {fields}")     # ['Suite A', '199', '150']

# 2. join() - Combine a list into a string
# Syntax: "separator".join(list)
# NOTE: The separator goes BEFORE .join(), not inside!
regions = ["EMEA", "AMER", "APAC"]
csv = ",".join(regions)        # Join with comma
print(f"CSV: {csv}")           # "EMEA,AMER,APAC"

# Common pattern: split → process → join
# Example: Convert "a-b-c" to "A|B|C"
# parts = "a-b-c".split("-")   # ['a', 'b', 'c']
# upper_parts = [p.upper() for p in parts]  # ['A', 'B', 'C']
# result = "|".join(upper_parts)  # "A|B|C"
```
