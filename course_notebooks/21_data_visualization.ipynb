## 2. Line Charts - Showing Trends

**Use line charts for:**
- Time series data
- Showing trends
- Changes over time

### Demo 2.1: Basic Line Chart

```python
# Demo: Basic Line Chart
# Line charts show trends over time
# Syntax: df.plot(x='column', y='column', kind='line')

# Monthly revenue data
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
revenue = [15000, 18000, 16500, 19000, 17500, 21000]

df = pd.DataFrame({'Month': months, 'Revenue': revenue})

# Create line chart using pandas .plot()
df.plot(x='Month', y='Revenue', kind='line')

# Add title and labels
plt.title('Monthly Revenue Trend')
plt.ylabel('Revenue (€)')

# Display the chart
plt.show()
```

### Demo 2.2: Multiple Lines

```python
# Demo: Multiple Lines for Comparison
# Plot multiple columns on the same chart
# Syntax: df.plot(x='col', y=['col1', 'col2'], kind='line')

# Compare two products over time
df = pd.DataFrame({
    'Month': months,
    'Suite A': [15000, 18000, 16500, 19000, 17500, 21000],
    'Suite B': [8500, 9200, 8800, 10500, 9800, 11200]
})

# Plot both products - pass a list of column names to y
df.plot(x='Month', y=['Suite A', 'Suite B'], kind='line', marker='o')

# Customize the chart
plt.title('Product Revenue Comparison')
plt.ylabel('Revenue (€)')
plt.legend(title='Product')      # Add legend title
plt.grid(True, alpha=0.3)        # Subtle grid lines
plt.show()
```

## 3. Bar Charts - Comparing Categories

**Use bar charts for:**
- Comparing different categories
- Ranking items
- Showing composition

### Demo 3.1: Vertical Bar Chart

```python
# Demo: Vertical Bar Chart
# Bar charts compare categories
# Syntax: df.plot(x='category', y='value', kind='bar')

# Regional sales data
regions = ['EMEA', 'AMER', 'APAC']
sales = [350000, 280000, 245000]

df = pd.DataFrame({'Region': regions, 'Sales': sales})

# Create vertical bar chart
df.plot(x='Region', y='Sales', kind='bar', color='steelblue')

# Customize
plt.title('Sales by Region')
plt.ylabel('Total Sales (€)')
plt.xticks(rotation=0)  # Keep labels horizontal
plt.show()
```

### Demo 3.2: Horizontal Bar Chart

```python
# Demo: Horizontal Bar Chart
# Use horizontal bars for long category names or rankings
# Syntax: kind='barh' (bar horizontal)

# Top 5 products - sorted for visual clarity
products = ['Analytics', 'Suite A', 'Suite B', 'Premium', 'Basic']
revenue = [185000, 165000, 142000, 128000, 95000]

df = pd.DataFrame({'Product': products, 'Revenue': revenue})
df = df.sort_values('Revenue')  # Sort ascending so highest is at top

# Horizontal bar chart
df.plot(x='Product', y='Revenue', kind='barh', color='coral')

plt.title('Top 5 Products by Revenue')
plt.xlabel('Revenue (€)')
plt.show()
```

### Demo 3.3: Grouped Bar Chart

```python
# Demo: Grouped Bar Chart
# Compare multiple values across categories
# Plot multiple columns without specifying y (plots all numeric columns)

# Products across regions
df = pd.DataFrame({
    'Region': ['EMEA', 'AMER', 'APAC'],
    'Suite A': [95000, 78000, 68000],
    'Suite B': [62000, 52000, 48000],
    'Analytics': [85000, 72000, 65000]
})

# Set index to Region, then plot all other columns
df.plot(x='Region', kind='bar', width=0.8)

plt.title('Product Performance by Region')
plt.ylabel('Revenue (€)')
plt.xticks(rotation=0)
plt.legend(title='Product')
plt.show()
```

## 4. Scatter Plots - Showing Relationships

**Use scatter plots for:**
- Relationships between two variables
- Correlations
- Finding patterns or outliers

### Demo 4.1: Basic Scatter Plot

```python
# Demo: Scatter Plot
# Scatter plots show relationship between two numeric variables
# Syntax: df.plot(kind='scatter', x='col1', y='col2')

# Marketing spend vs Revenue - is there a correlation?
df = pd.DataFrame({
    'Marketing': [5000, 8000, 6500, 12000, 9500, 15000, 11000, 7500],
    'Revenue': [45000, 68000, 52000, 95000, 78000, 118000, 88000, 62000]
})

# Create scatter plot
# s = marker size, alpha = transparency
df.plot(kind='scatter', x='Marketing', y='Revenue', s=100, alpha=0.6)

plt.title('Marketing Spend vs Revenue')
plt.xlabel('Marketing Spend (€)')
plt.ylabel('Revenue (€)')
plt.grid(True, alpha=0.3)
plt.show()

print("Pattern: More marketing → Higher revenue!")
```

### Demo 4.2: Colored Scatter Plot

```python
# Demo: Colored Scatter Plot by Category
# Use different colors for different groups
# Requires matplotlib directly for color grouping

import numpy as np

df = pd.DataFrame({
    'Price': [99, 199, 299, 499, 149, 249, 399, 599],
    'Units': [850, 520, 280, 120, 640, 380, 190, 85],
    'Region': ['EMEA', 'EMEA', 'EMEA', 'EMEA', 'AMER', 'AMER', 'AMER', 'AMER']
})

# Create separate scatter for each region
# Loop through unique regions and plot each with different color
for region in df['Region'].unique():
    subset = df[df['Region'] == region]
    plt.scatter(subset['Price'], subset['Units'], label=region, s=100, alpha=0.6)

plt.title('Price vs Units Sold by Region')
plt.xlabel('Price (€)')
plt.ylabel('Units Sold')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 5. Histograms - Showing Distributions

**Use histograms for:**
- Understanding data distribution
- Finding central tendency
- Identifying skewness

### Demo 5.1: Transaction Amount Distribution

```python
# Demo: Histogram - Distribution of Values
# Histograms show how values are distributed
# Syntax: df['column'].plot(kind='hist', bins=n)

# Generate sample transaction data
np.random.seed(42)
transactions = np.random.normal(2500, 800, 200)  # Mean €2500, std deviation €800

df = pd.DataFrame({'Amount': transactions})

# Create histogram
# bins = number of bars (groups)
# edgecolor adds borders to bars
df['Amount'].plot(kind='hist', bins=20, color='skyblue', edgecolor='black')

plt.title('Transaction Amount Distribution')
plt.xlabel('Transaction Amount (€)')
plt.ylabel('Frequency')

# Add vertical line at mean
plt.axvline(df['Amount'].mean(), color='red', linestyle='--', linewidth=2, label='Mean')
plt.legend()
plt.show()

print(f"Mean: €{df['Amount'].mean():,.2f}")
print(f"Most transactions are between €2000-€3000")
```

## 6. Box Plots - Statistical Summaries

**Use box plots for:**
- Comparing distributions
- Finding outliers
- Showing quartiles

### Demo 6.1: Regional Revenue Distribution

```python
# Demo: Box Plot - Statistical Summary
# Box plots show median, quartiles, and outliers
# Syntax: df.plot(kind='box')

# Revenue data by region
np.random.seed(42)
df = pd.DataFrame({
    'EMEA': np.random.normal(3000, 500, 50),  # Higher mean
    'AMER': np.random.normal(2500, 600, 50),  # Medium mean, more spread
    'APAC': np.random.normal(2200, 400, 50)   # Lower mean, less spread
})

# Create box plot - one box per column
df.plot(kind='box')

plt.title('Revenue Distribution by Region')
plt.ylabel('Transaction Amount (€)')
plt.show()

print("Box shows: median (line), 25th-75th percentile (box), outliers (dots)")
```

## 8. Multi-Chart Dashboards

Create multiple charts in one figure.

### Demo 8.1: 2×2 Dashboard

```python
# Demo: 2×2 Dashboard - Multiple Charts in One Figure
# plt.subplots() creates a grid of charts
# Syntax: fig, axes = plt.subplots(rows, cols, figsize=(width, height))

# Sample data
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
revenue = [15000, 18000, 16500, 19000, 17500, 21000]
regions = ['EMEA', 'AMER', 'APAC']
regional_sales = [350000, 280000, 245000]

# Create 2x2 grid of subplots
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Chart 1 (top-left): Revenue trend
# axes[row, col] selects the subplot
axes[0, 0].plot(months, revenue, marker='o', color='green', linewidth=2)
axes[0, 0].set_title('Monthly Revenue Trend', fontsize=14, fontweight='bold')
axes[0, 0].set_ylabel('Revenue (€)')
axes[0, 0].grid(True, alpha=0.3)

# Chart 2 (top-right): Regional sales bar
axes[0, 1].bar(regions, regional_sales, color=['steelblue', 'coral', 'lightgreen'])
axes[0, 1].set_title('Sales by Region', fontsize=14, fontweight='bold')
axes[0, 1].set_ylabel('Sales (€)')

# Chart 3 (bottom-left): Histogram
axes[1, 0].hist(np.random.normal(18000, 2000, 100), bins=15, color='skyblue', edgecolor='black')
axes[1, 0].set_title('Revenue Distribution', fontsize=14, fontweight='bold')
axes[1, 0].set_xlabel('Revenue (€)')
axes[1, 0].set_ylabel('Frequency')

# Chart 4 (bottom-right): Scatter plot
prices = [99, 199, 299, 499]
units = [850, 520, 280, 120]
axes[1, 1].scatter(prices, units, s=200, alpha=0.6, color='purple')
axes[1, 1].set_title('Price vs Units Sold', fontsize=14, fontweight='bold')
axes[1, 1].set_xlabel('Price (€)')
axes[1, 1].set_ylabel('Units')
axes[1, 1].grid(True, alpha=0.3)

# Add overall title
plt.suptitle('Q2 2024 Sales Dashboard', fontsize=16, fontweight='bold', y=0.995)

# Adjust layout to prevent overlap
plt.tight_layout()
plt.show()
```

## Challenge Exercise: Complete Visual Report

Create a comprehensive 4-chart dashboard:

Given annual sales data:
1. Line chart: Monthly trend
2. Bar chart: Top 5 products
3. Scatter plot: Marketing spend vs Revenue
4. Box plot: Revenue by quarter

All in one figure with proper titles and labels.

```python
# Prepare data
np.random.seed(42)

# Monthly data
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
monthly_revenue = [15000 + i*1200 + np.random.randint(-1000, 2000) for i in range(12)]

# Product data
products = ['Analytics', 'Suite A', 'Suite B', 'Premium', 'Basic']
product_revenue = [185000, 165000, 142000, 128000, 95000]

# Marketing data
marketing = [5000 + i*1000 for i in range(12)]

# Quarterly data
q1 = np.random.normal(16000, 1500, 25)
q2 = np.random.normal(18500, 1600, 25)
q3 = np.random.normal(20000, 1700, 25)
q4 = np.random.normal(22500, 1800, 25)

# Your solution here - create 2x2 dashboard
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# Chart 1: Monthly trend
axes[0, 0].plot(months, monthly_revenue, marker='o', linewidth=2, color='green')
axes[0, 0].set_title('2024 Monthly Revenue Trend', fontsize=14, fontweight='bold')
axes[0, 0].set_ylabel('Revenue (€)')
axes[0, 0].tick_params(axis='x', rotation=45)
axes[0, 0].grid(True, alpha=0.3)

# Chart 2: Top products
axes[0, 1].barh(products, product_revenue, color='coral')
axes[0, 1].set_title('Top 5 Products by Revenue', fontsize=14, fontweight='bold')
axes[0, 1].set_xlabel('Revenue (€)')

# Chart 3: Marketing vs Revenue
axes[1, 0].scatter(marketing, monthly_revenue, s=150, alpha=0.6, color='purple')
axes[1, 0].set_title('Marketing Spend vs Revenue', fontsize=14, fontweight='bold')
axes[1, 0].set_xlabel('Marketing Spend (€)')
axes[1, 0].set_ylabel('Revenue (€)')
axes[1, 0].grid(True, alpha=0.3)

# Chart 4: Quarterly box plots
quarterly_data = pd.DataFrame({
    'Q1': q1,
    'Q2': q2,
    'Q3': q3,
    'Q4': q4
})
quarterly_data.plot(kind='box', ax=axes[1, 1])
axes[1, 1].set_title('Quarterly Revenue Distribution', fontsize=14, fontweight='bold')
axes[1, 1].set_ylabel('Revenue (€)')

plt.suptitle('2024 Annual Sales Analysis Dashboard', fontsize=18, fontweight='bold', y=0.995)
plt.tight_layout()
plt.show()

print("\n📊 Dashboard complete! All key metrics visualized.")
```
