## 2. How to Ask Good Questions

### ❌ Bad Questions:
- "Help me with Pandas"
- "My code doesn't work"
- "Fix this"

### ✅ Good Questions:
- "How do I filter a Pandas DataFrame to keep only rows where Revenue > 1000?"
- "I'm getting a KeyError on this line: df['Sales']. Here's my DataFrame structure: [paste head()]. What's wrong?"
- "What's the difference between merge() and concat() in Pandas? When should I use each?"

### Framework for Good Questions:

1. **Context**: What are you trying to achieve?
2. **Current situation**: What have you tried?
3. **Problem**: What's not working?
4. **Data structure**: What does your data look like?
5. **Expected outcome**: What do you want to happen?

**Example:**
```
I'm analyzing sales data in Pandas. I have a DataFrame with columns: Date, Product, Revenue.
I want to calculate monthly total revenue.
I tried: df.groupby('Date')['Revenue'].sum()
But it's grouping by day, not month.
How do I group by month instead?
```

## 3. Common Use Cases for AI

### Use Case 1: Explaining Concepts

````python
# Use Case 1: Asking AI to Explain Code
# This is one of the BEST uses of AI - understanding code you encounter

# You see this code and don't understand it:
result = df.groupby('Region')['Revenue'].agg(['sum', 'mean'])

# How to ask AI effectively:
# "Can you explain this line of code step by step?
#  df.groupby('Region')['Revenue'].agg(['sum', 'mean'])
#  I understand groupby, but what does .agg(['sum', 'mean']) do?"

# AI will explain:
# 1. df.groupby('Region') - splits data by unique Region values
# 2. ['Revenue'] - selects the Revenue column from each group
# 3. .agg(['sum', 'mean']) - applies BOTH sum and mean to each group
# 4. Result: DataFrame with Region as index, sum and mean as columns
````

### Use Case 2: Generating Example Code

````python
# Use Case 2: Ask AI to Generate Example Code
# When you know WHAT you want but not HOW to do it

# Good prompt to AI:
# "Show me an example of how to create a pivot table in Pandas
#  with regions as rows and products as columns, showing total revenue.
#  Use simple sample data."

# AI will generate working code with sample data you can learn from
# This is faster than reading documentation for basic examples
# BUT: Always understand the code before using it!
````

### Use Case 3: Debugging Errors

````python
# Use Case 3: Ask AI to Help Debug Errors
# Include the FULL error message and relevant context

# You get an error:
# KeyError: 'Sales'

# Bad question: "My code doesn't work"

# Good question to AI:
# "I'm getting KeyError: 'Sales' on this line:
#  total = df['Sales'].sum()
# 
#  My DataFrame columns are: ['Date', 'Product', 'Revenue', 'Quantity']
#  What's wrong and how do I fix it?"

# AI will point out:
# - 'Sales' doesn't exist in your columns
# - You probably meant 'Revenue'
# - Use df.columns to check available column names
````

### Use Case 4: Finding Better Solutions

````python
# Use Case 4: Ask AI for Better Solutions
# AI can suggest more "Pythonic" or efficient approaches

# Your current approach (works but inefficient):
results = []
for region in df['Region'].unique():
    subset = df[df['Region'] == region]
    total = subset['Revenue'].sum()
    results.append({'Region': region, 'Total': total})

# Good question to AI:
# "I'm using a for loop to calculate total revenue by region.
#  Is there a more efficient Pandas way to do this?
#  Here's my current code: [paste code above]"

# AI will suggest the Pandas way (single line!):
# results = df.groupby('Region')['Revenue'].sum()

# Learning: Pandas is designed to avoid loops when possible
# groupby is almost always faster than iterating manually
````

## 4. Understanding AI-Generated Code

### ⚠️ NEVER Blindly Copy Code!

**Always:**
1. Read the code line by line
2. Understand what each part does
3. Test with your actual data
4. Verify the output is correct
5. Ask follow-up questions if unclear

````python
# CRITICAL: Always Understand AI-Generated Code
# Never blindly copy-paste - understand what each part does!

# Example: AI suggests this code
df_sorted = df.sort_values(['Region', 'Revenue'], ascending=[True, False])

# Before using it, break it down:
# - sort_values() = sorts the DataFrame rows
# - ['Region', 'Revenue'] = sort by Region first, then by Revenue (within each Region)
# - ascending=[True, False] = Region alphabetically (A→Z), Revenue descending (high→low)

# Always:
# 1. Read the code line by line
# 2. Understand what each parameter does
# 3. Test with your actual data (not just the example)
# 4. Verify the output is what you expected
# 5. Ask follow-up questions if anything is unclear
````

## 5. When to Use AI vs Documentation

### Use AI for:
- Quick explanations
- "How do I...?" questions
- Debugging specific errors
- Comparing approaches
- Learning concepts

### Use Official Documentation for:
- Complete parameter lists
- Official behavior definitions
- Edge cases
- Version-specific features
- Authoritative answers

### Best Approach:
1. Ask AI to explain concept
2. Check documentation for details
3. Test with your data
4. Ask AI for clarifications if needed

## 6. AI Limitations & Pitfalls

### Common AI Mistakes:

1. **Hallucination** - Making up functions that don't exist
   ```python
   # AI might suggest (doesn't exist!):
   df.smart_analyze()  # Not a real Pandas method!
   ```

2. **Outdated Syntax** - Using old library versions
   ```python
   # Old way AI might suggest:
   df.append(new_row)  # Deprecated in newer Pandas!
   # Better: pd.concat([df, new_row])
   ```

3. **Doesn't Know Your Data** - Assumes column names
   ```python
   # AI assumes you have 'Sales' column
   df['Sales'].sum()  # But your column is 'Revenue'!
   ```

4. **Overcomplicated Solutions** - Not always the simplest
   
5. **Security Issues** - May suggest unsafe practices
