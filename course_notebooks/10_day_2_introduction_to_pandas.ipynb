## 1. What is Pandas?

**Pandas** is Python's premier data analysis library. Think of it as **"Excel for Python"** - but much more powerful!

### Why Use Pandas?

| Task | Without Pandas | With Pandas |
|------|---------------|-------------|
| Read a CSV file | 10+ lines of code | 1 line |
| Calculate column sum | Loop through all rows | 1 line |
| Filter data | Complex list comprehensions | Simple syntax |
| Group and aggregate | Manual dictionary work | Built-in methods |

### The Core Concept: DataFrame

A **DataFrame** is a table with:
- **Rows**: Individual records (like rows in Excel)
- **Columns**: Data fields (like columns in Excel)
- **Index**: Row labels (like row numbers in Excel)

## 3. Creating DataFrames

There are several ways to create a DataFrame. The most common are from **dictionaries** and from **files**.

### Demo 3.1: DataFrame from Dictionary of Lists

This is the most common way to create a DataFrame manually:

````python
# Demo: Create DataFrame from Dictionary of Lists
# This is the most common way to create a DataFrame manually
# Structure: Each dictionary KEY becomes a COLUMN NAME
#            Each dictionary VALUE (a list) becomes the COLUMN DATA

import pandas as pd

# Step 1: Create a dictionary with your data
# Keys = column names (strings)
# Values = lists of data (all lists must be same length!)
data = {
    "Product": ["Suite A", "Suite B", "Suite C", "Analytics"],  # String column
    "Price": [199, 99, 299, 399],                               # Integer column
    "Units": [150, 200, 80, 120]                                # Integer column
}

# Step 2: Create DataFrame using pd.DataFrame()
# Syntax: pd.DataFrame(dictionary)
df = pd.DataFrame(data)

# Step 3: Display the DataFrame
# Just typing the variable name or using print() shows the table
print(df)

# Check the type - it should be pandas.core.frame.DataFrame
print(f"\nType: {type(df)}")
````

**Notice:**
- The left side (0, 1, 2, 3) is the **index** - row labels
- Column names come from dictionary keys
- Data is aligned in a nice table format

### Demo 3.2: DataFrame from List of Dictionaries

This format is common when data comes from APIs or databases:

````python
# Demo: Create DataFrame from List of Dictionaries
# Alternative structure: each dictionary represents ONE ROW
# This format is common when data comes from APIs, databases, or JSON files
# Structure: List of dicts where each dict has the same keys

# Each dictionary is one row of data
# Dictionary keys become column names
records = [
    {"Region": "EMEA", "Revenue": 125000, "Employees": 45},  # Row 0
    {"Region": "AMER", "Revenue": 98000, "Employees": 38},   # Row 1
    {"Region": "APAC", "Revenue": 67000, "Employees": 25}    # Row 2
]

# Create DataFrame - pandas figures out the structure automatically
# Syntax: pd.DataFrame(list_of_dictionaries)
df_regions = pd.DataFrame(records)

print(df_regions)
````

## 4. Loading Data from CSV Files

In real work, you'll usually load data from files. **CSV (Comma-Separated Values)** is the most common format.

Pandas makes this incredibly simple!

### Demo 4.1: Reading a CSV File

````python
# Demo: Reading a CSV File
# CSV = Comma-Separated Values, the most common data file format
# pd.read_csv() loads a CSV file into a DataFrame in one line!
# Syntax: df = pd.read_csv("filename.csv")

# First, let's create a sample CSV file to work with
# (In real work, you'd already have the CSV file)
sample_data = pd.DataFrame({
    "Date": ["2024-01-15", "2024-01-16", "2024-01-17", "2024-01-18", "2024-01-19"],
    "Product": ["Suite A", "Suite B", "Suite A", "Analytics", "Suite B"],
    "Region": ["EMEA", "AMER", "APAC", "EMEA", "AMER"],
    "Revenue": [1500, 990, 1500, 3990, 990]
})

# Save to CSV file using .to_csv()
# index=False prevents writing row numbers as a column
sample_data.to_csv("sample_sales.csv", index=False)
print("Sample CSV created!")

# NOW read the CSV file back into a new DataFrame
# This is what you'll do with real data files
df = pd.read_csv("sample_sales.csv")

print("\nLoaded from CSV:")
print(df)
````

### Demo 4.2: Common read_csv Options

````python
# Demo: Common read_csv Options
# pd.read_csv() has many optional parameters to handle different file formats

# Basic read (most common - just the filename)
df = pd.read_csv("sample_sales.csv")

# Full syntax with common options:
# df = pd.read_csv(
#     "filename.csv",           # Required: path to the file
#     sep=",",                  # Column separator (comma is default)
#     header=0,                 # Which row contains column names (0 = first row)
#     encoding="utf-8",         # Character encoding (utf-8 handles most files)
#     parse_dates=["Date"],     # List of columns to convert to datetime
# )

# Useful option: parse_dates converts date columns automatically
# Without it, dates are loaded as strings
# With it, dates become datetime objects for date calculations
print("Read with parse_dates to convert Date column:")
df = pd.read_csv("sample_sales.csv", parse_dates=["Date"])

# Check data types - Date should now be datetime64 instead of object
print(df.dtypes)
````

## 5. Exploring Your Data

When you load a new dataset, **always explore it first!** These are the essential commands.

### Demo 5.1: First Look at Your Data

````python
# Demo: Create a Larger Sample Dataset for Exploration
# We'll use this DataFrame to demonstrate all the exploration methods

import pandas as pd

# Create a 20-row dataset with various columns
# pd.date_range() creates a sequence of dates
# The * operator repeats a list (["A", "B"] * 2 = ["A", "B", "A", "B"])
sales = pd.DataFrame({
    "Date": pd.date_range("2024-01-01", periods=20, freq="D"),  # 20 consecutive days
    "Product": ["Suite A", "Suite B", "Analytics", "Suite C"] * 5,  # Repeats 5 times
    "Region": ["EMEA", "AMER", "APAC", "EMEA", "AMER"] * 4,  # Repeats 4 times
    "Revenue": [1500, 990, 3990, 2990, 1500, 990, 3990, 2990, 
                1650, 890, 4200, 3100, 1400, 1100, 3800, 2800,
                1550, 950, 4100, 3000],
    "Units": [10, 15, 5, 8, 12, 18, 6, 9, 11, 14, 4, 7, 13, 16, 5, 8, 10, 17, 6, 9]
})

print("Our dataset:")
print(sales)
````

### Demo 5.2: head() and tail()

````python
# Demo: head() - View the First Rows
# Essential for quickly checking if data loaded correctly
# Shows first N rows (default is 5)
# Syntax: df.head() or df.head(n)

# Default: show first 5 rows
print("First 5 rows (default):")
print(sales.head())

# Specify number of rows to show
print("\nFirst 3 rows:")
print(sales.head(3))
````

````python
# Demo: tail() - View the Last Rows
# Useful for checking the end of your data
# Shows last N rows (default is 5)
# Syntax: df.tail() or df.tail(n)

# Default: show last 5 rows
print("Last 5 rows (default):")
print(sales.tail())

# Specify number of rows to show
print("\nLast 2 rows:")
print(sales.tail(2))
````

### Demo 5.3: shape - How Big Is My Data?

````python
# Demo: shape - Get DataFrame Dimensions
# Returns a tuple: (number_of_rows, number_of_columns)
# Note: shape is an ATTRIBUTE, not a method - no parentheses!
# Syntax: df.shape

# Get both dimensions at once
print(f"Shape: {sales.shape}")

# Access individual dimensions using index
# shape[0] = number of rows
# shape[1] = number of columns
print(f"Rows: {sales.shape[0]}")
print(f"Columns: {sales.shape[1]}")

# Alternative: len() returns just the row count
print(f"\nNumber of rows (using len): {len(sales)}")
````

### Demo 5.4: columns - What Columns Exist?

````python
# Demo: columns - Get Column Names
# Returns an Index object containing all column names
# Note: columns is an ATTRIBUTE, not a method - no parentheses!
# Syntax: df.columns

# Get column names (returns pandas Index object)
print("Columns:")
print(sales.columns)

# Convert to a regular Python list using .tolist()
# Useful when you need to iterate or manipulate column names
print("\nAs list:")
print(sales.columns.tolist())
````

### Demo 5.5: dtypes - What Are the Data Types?

````python
# Demo: dtypes - Check Data Types of Each Column
# Shows the data type pandas assigned to each column
# CRITICAL for understanding your data and avoiding errors!
# Syntax: df.dtypes (attribute, no parentheses)

print("Data types:")
print(sales.dtypes)

# Common pandas data types:
# int64      - Integers (whole numbers)
# float64    - Decimals (numbers with decimal points)
# object     - Strings (text) - NOTE: pandas calls strings "object"
# datetime64 - Dates and times
# bool       - Boolean (True/False)
# category   - Categorical data (limited set of values)
````

### Demo 5.6: info() - Complete Summary

````python
# Demo: info() - Complete DataFrame Summary
# The MOST USEFUL exploration method - shows everything at once!
# Syntax: df.info() (method with parentheses)

print("DataFrame Info:")
sales.info()

# info() displays:
# - Class type (DataFrame)
# - RangeIndex: total number of entries (rows)
# - Data columns: number of columns
# - For each column:
#   - Column name
#   - Non-Null Count: how many non-missing values (helps spot missing data!)
#   - Dtype: data type
# - Memory usage
#
# TIP: If Non-Null Count < total rows, you have missing values!
````

### Demo 5.7: describe() - Statistical Summary

````python
# Demo: describe() - Statistical Summary of Numeric Columns
# Calculates key statistics for all numeric columns at once
# Syntax: df.describe()

print("Statistical Summary:")
print(sales.describe())

# describe() shows (for each numeric column):
# count - Number of non-null values
# mean  - Average (sum / count)
# std   - Standard deviation (spread of data)
# min   - Minimum value
# 25%   - First quartile (25th percentile)
# 50%   - Median (middle value)
# 75%   - Third quartile (75th percentile)
# max   - Maximum value
#
# NOTE: Only shows numeric columns by default
# For text columns, use: df.describe(include='object')
````

## 6. Selecting Columns

Often you need to work with specific columns, not the entire DataFrame.

### Demo 6.1: Select a Single Column

````python
# Demo: Select a Single Column
# Use bracket notation with the column name as a string
# Syntax: df["column_name"]
# Returns a Series (one-dimensional pandas object)

# Select the Revenue column
revenue = sales["Revenue"]

print("Revenue column:")
print(revenue)

# Check the type - single column returns a SERIES, not a DataFrame
# Series = one column of data with an index
print(f"\nType: {type(revenue)}")
````

````python
# Demo: Common Operations on a Single Column
# Once you select a column, you can call aggregation methods on it
# These methods return a single value summarizing the column

# Syntax: df['column_name'].method()

# .sum() - Add up all values
print(f"Total Revenue: €{sales['Revenue'].sum():,.2f}")

# .mean() - Calculate the average
print(f"Average Revenue: €{sales['Revenue'].mean():,.2f}")

# .max() - Find the highest value
print(f"Max Revenue: €{sales['Revenue'].max():,.2f}")

# .min() - Find the lowest value
print(f"Min Revenue: €{sales['Revenue'].min():,.2f}")
````

### Demo 6.2: Select Multiple Columns

````python
# Demo: Select Multiple Columns
# Use DOUBLE brackets with a LIST of column names
# Syntax: df[["col1", "col2", "col3"]]
# Returns a DataFrame (not a Series)

# Select Product and Revenue columns
# The inner brackets create a list: ["Product", "Revenue"]
# The outer brackets select from the DataFrame
subset = sales[["Product", "Revenue"]]

print("Product and Revenue columns:")
print(subset.head())

# Check the type - multiple columns returns a DATAFRAME
print(f"\nType: {type(subset)}")
````

````python
# Demo: Understanding Double vs Single Brackets
# This is a common point of confusion!

# SINGLE brackets + string → Series (one column)
# df["col"] → Series

# DOUBLE brackets + list → DataFrame (can be one or more columns)
# df[["col1", "col2"]] → DataFrame
# df[["col"]] → DataFrame (even with one column!)

# Why double brackets work:
# Step 1: Inner brackets create a Python list: ["Product", "Region", "Revenue"]
# Step 2: Outer brackets pass that list to the DataFrame selector

# You can also create the list first, then use single brackets:
column_list = ["Product", "Region", "Revenue"]
subset = sales[column_list]  # Same as sales[["Product", "Region", "Revenue"]]
print(subset.head())
````

### Demo 6.3: Common Column Operations

````python
# Demo: Common Column Operations
# Useful methods for exploring categorical/text columns

# .unique() - Get all distinct values in a column
# Returns an array of unique values
print("Unique products:")
print(sales["Product"].unique())

# .nunique() - Count how many unique values exist
# Returns a single integer
print(f"\nNumber of unique products: {sales['Product'].nunique()}")

# .value_counts() - Count frequency of each value
# Returns a Series with values as index and counts as data
# Sorted by count (most frequent first)
print("\nProduct frequency:")
print(sales["Product"].value_counts())
````

## 7. Adding and Calculating Columns

You can create new columns based on existing data.

### Demo 7.1: Add a Calculated Column

````python
# Demo: Add a Calculated Column
# Create new columns by performing operations on existing columns
# Syntax: df["new_column"] = calculation

# Create a fresh copy to work with
# .copy() prevents changes from affecting the original
df = sales.copy()

# Add a new column: Unit Price = Revenue / Units
# Pandas performs the division element-by-element (row by row)
df["UnitPrice"] = df["Revenue"] / df["Units"]

# Display selected columns to see the result
print("With UnitPrice column:")
print(df[["Product", "Revenue", "Units", "UnitPrice"]].head())
````

````python
# Demo: Multiple Calculated Columns
# You can add as many calculated columns as needed
# Operations work on entire columns at once (vectorized operations)

# Add revenue with 20% tax
# Multiply entire column by 1.20 (100% + 20%)
df["RevenueWithTax"] = df["Revenue"] * 1.20

# Add revenue with 10% discount
# Multiply by 0.90 (100% - 10%)
df["DiscountedRevenue"] = df["Revenue"] * 0.90

# Display the results
print("With calculated columns:")
print(df[["Revenue", "RevenueWithTax", "DiscountedRevenue"]].head())
````

### Demo 7.2: Column Arithmetic

````python
# Demo: Building Profit Calculations
# Common business calculation pattern: Revenue → Cost → Profit → Margin

# Step 1: Create a Cost column (assume 60% of revenue is cost)
df["Cost"] = df["Revenue"] * 0.6

# Step 2: Calculate Profit (Revenue minus Cost)
df["Profit"] = df["Revenue"] - df["Cost"]

# Step 3: Calculate Profit Margin percentage
# Formula: (Profit / Revenue) * 100
df["ProfitMargin"] = (df["Profit"] / df["Revenue"]) * 100

# Display the financial columns together
print("Profit calculations:")
print(df[["Revenue", "Cost", "Profit", "ProfitMargin"]].head())
````

## 8. Basic Filtering (Preview)

You can filter rows based on conditions. We'll cover this in depth tomorrow, but here's a preview:

````python
# Demo: Basic Filtering (Preview)
# Filter rows based on a condition
# Syntax: df[df["column"] condition value]
# We'll cover this in depth on Day 3!

# Filter: Keep only rows where Revenue is greater than 2000
# The condition df["Revenue"] > 2000 creates a True/False series
# Passing that series to df[] keeps only True rows
high_revenue = sales[sales["Revenue"] > 2000]

print("Transactions with Revenue > €2000:")
print(high_revenue)

# Check how many rows matched the filter
print(f"\nCount: {len(high_revenue)} rows")
````

````python
# Demo: Filter for Specific Values
# Use == for exact match (note: double equals for comparison!)
# Syntax: df[df["column"] == "value"]

# Filter: Keep only EMEA region rows
emea_sales = sales[sales["Region"] == "EMEA"]

print("EMEA Sales:")
print(emea_sales)

# Calculate total for filtered data
# Chain the filter with aggregation
print(f"\nEMEA Total: €{emea_sales['Revenue'].sum():,.2f}")
````
