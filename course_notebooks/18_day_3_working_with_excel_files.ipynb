## 1. Reading Excel FilesPandas makes Excel easy.

```python
# Demo: Reading Excel Files
# pd.read_excel() loads Excel files (.xlsx, .xls) into DataFrames
# Works just like read_csv() but for Excel files
# Syntax: df = pd.read_excel("filename.xlsx")

import pandas as pd

# Basic Excel read (reads first sheet by default)
df = pd.read_excel("sales_data.xlsx")
print(df.head())

# Read a SPECIFIC sheet by name
# Use sheet_name parameter
df = pd.read_excel("sales_data.xlsx", sheet_name="Q1")
print(df)
```

## 2. Writing to ExcelExport your analysis results.

```python
# Demo: Writing DataFrames to Excel
# .to_excel() exports a DataFrame to an Excel file
# Syntax: df.to_excel("filename.xlsx", index=False, sheet_name="name")

import pandas as pd

# Create sample data
df = pd.DataFrame({
    "Region": ["EMEA", "AMER", "APAC"],
    "Revenue": [125000, 98000, 67000]
})

# Write to Excel
# index=False prevents row numbers from being written as a column
# sheet_name sets the name of the worksheet tab
df.to_excel("output.xlsx", index=False, sheet_name="Sales Summary")
print("Excel file created successfully")
```

## 3. Multiple SheetsCreate Excel files with multiple sheets.

```python
# Demo: Creating Excel Files with Multiple Sheets
# Use pd.ExcelWriter as a context manager to write multiple sheets
# Syntax: with pd.ExcelWriter("file.xlsx") as writer:
#             df1.to_excel(writer, sheet_name="Sheet1")
#             df2.to_excel(writer, sheet_name="Sheet2")

# Use 'with' statement - automatically saves and closes the file
with pd.ExcelWriter("report.xlsx") as writer:
    # Write same data to different sheets (or use different DataFrames)
    df.to_excel(writer, sheet_name="Summary", index=False)
    df.to_excel(writer, sheet_name="Details", index=False)

print("Multi-sheet Excel file created")
```
