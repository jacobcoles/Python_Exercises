## 1. Basic If Statements

The `if` statement lets your code make decisions based on conditions.

**Syntax:**
```python
if condition:
    # code to run if condition is True
```

**Important:** Python uses indentation (4 spaces) to define code blocks!

### Demo: Simple If Statements

````python
# Demo 1: Basic if statement
# An 'if' statement checks a condition and runs code if it's True
# Syntax: if condition:
#             indented code block (runs only if True)

revenue = 125000

# The condition revenue > 100000 evaluates to True or False
# If True, the indented code below runs
# The colon : is required at the end of the if line
if revenue > 100000:
    # This indented block runs ONLY when condition is True
    # Python uses 4 spaces for indentation (very important!)
    print("High-value transaction - requires approval")
    print(f"Amount: €{revenue:,}")
````

````python
# Demo 2: If with else
# 'else' provides an alternative when the condition is False
# Syntax: if condition:
#             code if True
#         else:
#             code if False

amount = 4987.88

# Check if amount is greater than 5000
if amount > 5000:
    # Runs if condition is True
    print("Large transaction")
else:
    # Runs if condition is False (amount <= 5000)
    print("Standard transaction")
    
# Code after the if/else block runs regardless of the condition
# This line is NOT indented, so it always executes
print(f"Amount: €{amount:,.2f}")
````

````python
# Demo 3: Checking string equality
# Use == to compare if two values are equal
# Works with strings, numbers, and other types

status = "Active"

# Compare strings using == (double equals)
# Note: String comparison is case-sensitive ("Active" != "active")
if status == "Active":
    print("✓ Product is available for sale")
else:
    print("✗ Product is not available")
````

## 2. Elif for Multiple Conditions

Use `elif` (else if) when you have more than two possible outcomes.

**Syntax:**
```python
if condition1:
    # code for condition1
elif condition2:
    # code for condition2
else:
    # code if neither is true
```

### Demo: Multiple Conditions with Elif

````python
# Demo 1: Categorize transaction size using elif
# 'elif' (else if) allows checking multiple conditions in sequence
# Only the FIRST matching condition's code runs, then Python exits the chain

amount = 3500

# Python checks conditions top to bottom, stops at first True
if amount < 1000:
    # Runs if amount is less than 1000
    category = "Small"
elif amount < 5000:
    # Runs if first condition False AND amount is less than 5000
    # Since we get here, amount must be >= 1000 (first condition was False)
    category = "Medium"
elif amount < 10000:
    # Runs if previous conditions False AND amount < 10000
    category = "Large"
else:
    # Runs if ALL previous conditions were False
    # This means amount >= 10000
    category = "Very Large"

print(f"Amount: €{amount:,}")
print(f"Category: {category}")
````

````python
# Demo 2: Region-based logic using elif
# Use elif when checking one variable against multiple specific values
# This is like a "switch" or "case" statement in other languages

region_code = "EMEA"

# Check region and set appropriate currency and tax rate
if region_code == "AMER":
    currency = "USD"
    tax_rate = 0.07
elif region_code == "EMEA":
    currency = "EUR"
    tax_rate = 0.20
elif region_code == "APAC":
    currency = "USD"
    tax_rate = 0.10
else:
    # Default case: handle unknown regions
    currency = "USD"
    tax_rate = 0.00
    print("Warning: Unknown region, using defaults")

print(f"Region: {region_code}")
print(f"Currency: {currency}")
print(f"Tax Rate: {tax_rate:.0%}")  # :.0% formats as percentage with 0 decimals
````

````python
# Demo 3: Product tier pricing using elif
# Map product tier names to their prices
# Order doesn't matter when checking exact string matches

product_tier = "Pro"

# Check which tier and assign the corresponding price
if product_tier == "Basic":
    price = 49.00
elif product_tier == "Plus":
    price = 99.00
elif product_tier == "Pro":
    price = 199.00
elif product_tier == "Enterprise":
    price = 499.00
else:
    # Handle invalid tier - set price to 0 and show warning
    price = 0.00
    print(f"Unknown tier: {product_tier}")

print(f"Product: {product_tier}")
print(f"Price: €{price:.2f}")
````

## 3. Comparison Operators

Comparison operators let you compare values:
- `==` : Equal to
- `!=` : Not equal to
- `>` : Greater than
- `<` : Less than
- `>=` : Greater than or equal to
- `<=` : Less than or equal to

### Demo: Using Comparison Operators

````python
# Demo 1: All comparison operators
# Comparison operators compare two values and return True or False

a = 10
b = 5

print(f"a = {a}, b = {b}")

# == Equal to (is a equal to b?)
print(f"a == b: {a == b}")    # False (10 is not equal to 5)

# != Not equal to (is a different from b?)
print(f"a != b: {a != b}")    # True (10 is not equal to 5)

# > Greater than (is a bigger than b?)
print(f"a > b: {a > b}")      # True (10 is greater than 5)

# < Less than (is a smaller than b?)
print(f"a < b: {a < b}")      # False (10 is not less than 5)

# >= Greater than or equal to
print(f"a >= 10: {a >= 10}")  # True (10 is equal to 10)

# <= Less than or equal to
print(f"b <= 5: {b <= 5}")    # True (5 is equal to 5)
````

````python
# Demo 2: Comparing strings
# Strings can be compared with == and != just like numbers
# Comparison is CASE-SENSITIVE: "Active" != "active"

status = "Active"

# == checks if strings are exactly equal
if status == "Active":
    print("Product is active")

# != checks if strings are different
if status != "Discontinued":
    print("Product is still available")
````

````python
# Demo 3: Discount eligibility check
# Use >= to check "at least" conditions

units_ordered = 50
discount_threshold = 50

# >= means "greater than or equal to"
# True if units_ordered is 50 or more
if units_ordered >= discount_threshold:
    print(f"✓ Eligible for bulk discount")
    print(f"Ordered: {units_ordered} units (threshold: {discount_threshold})")
else:
    print(f"✗ Not eligible for bulk discount")
    print(f"Ordered: {units_ordered} units (need: {discount_threshold})")
````

## 4. Logical Operators: AND, OR, NOT

Combine multiple conditions:
- `and` : Both conditions must be True
- `or` : At least one condition must be True
- `not` : Reverses the boolean value

### Demo: AND Operator (Both Must Be True)

````python
# Demo 1: AND operator - BOTH conditions must be True
# 'and' requires ALL conditions to be True for the overall result to be True
# If ANY condition is False, the whole expression is False

revenue = 10000
customer_type = "Premium"

# Both conditions must be True:
# 1. revenue > 5000 → True (10000 > 5000)
# 2. customer_type == "Premium" → True
# True AND True = True
if revenue > 5000 and customer_type == "Premium":
    discount = 0.15
    print(f"✓ Approved: 15% premium discount")
    print(f"Revenue: €{revenue:,}, Customer: {customer_type}")
else:
    # Runs if EITHER condition is False
    discount = 0.05
    print(f"Standard 5% discount")
````

````python
# Demo 2: Product availability check using AND
# Product must be Active AND have stock to be available

status = "Active"
inventory = 25

# Both must be True:
# 1. status == "Active" → True
# 2. inventory > 0 → True (25 > 0)
if status == "Active" and inventory > 0:
    print("✓ Product available for sale")
    print(f"Status: {status}, Stock: {inventory}")
else:
    # Fails if product is not Active OR if inventory is 0
    print("✗ Product not available")
````

### Demo: OR Operator (At Least One Must Be True)

````python
# Demo 3: OR operator - At least ONE condition must be True
# 'or' requires only ONE condition to be True for the overall result to be True
# Only False if ALL conditions are False

amount = 50000
region = "LATAM"

# Either condition being True triggers the flag:
# 1. amount > 10000 → True (50000 > 10000)
# 2. region == "Unknown" → False
# True OR False = True
if amount > 10000 or region == "Unknown":
    print("⚠ Transaction requires manual review")
    print(f"Amount: €{amount:,}, Region: {region}")
else:
    print("Standard processing")
````

````python
# Demo 4: European region check using OR
# Check if region is ANY of the European regions

region_code = "EMEA"

# True if region is EMEA or NORD (both are European)
# Only one needs to match
if region_code == "EMEA" or region_code == "NORD":
    print("European region - apply GDPR rules")
    print(f"Region: {region_code}")
else:
    print("Non-European region")
````

### Demo: NOT Operator (Reverses the Condition)

````python
# Demo 5: NOT operator - reverses True/False
# 'not' flips the boolean value: not True = False, not False = True

category = "Legacy"

# 'not' inverts the condition result
# category == "Legacy" is True, so not True = False
if not category == "Legacy":
    print("Include in active product report")
else:
    print("Exclude from active product report")

# Cleaner alternative: use != instead of 'not =='
# This is more readable and preferred
if category != "Legacy":
    print("Include in active product report")
else:
    print("Exclude from active product report")
````

### Demo: Complex Business Rules

````python
# Demo 6: Multiple conditions combined with AND and OR
# Use parentheses () to group conditions and make logic clear

amount = 7500
region = "AMER"
product_category = "Suite"

# Complex logic with both AND and OR:
# (amount > 5000 AND region == "AMER") → (True AND True) → True
# OR
# product_category == "Suite" → True
# True OR True = True
if (amount > 5000 and region == "AMER") or product_category == "Suite":
    print("Requires senior manager approval")
    print(f"Amount: €{amount:,}, Region: {region}, Category: {product_category}")
elif amount > 10000:
    print("Requires director approval")
else:
    print("Standard approval process")

# TIP: Always use parentheses to make complex conditions readable
````

## 5. Nested Conditions

You can put if statements inside other if statements for more complex logic.

### Demo: Nested If Statements

````python
# Demo 1: Nested approval workflow
# You can put if statements inside other if statements
# The inner if only runs when the outer if is True

amount = 8000
is_approved = True

# Outer condition: check if high-value
if amount > 5000:
    print("High-value transaction")
    
    # Inner condition: only checked when outer is True
    # This is indented inside the outer if block
    if is_approved:
        print("✓ Approved for processing")
    else:
        print("✗ Pending approval")
else:
    # This runs when outer condition (amount > 5000) is False
    print("Standard transaction - auto-approved")
````

````python
# Demo 2: Customer tier and order size - nested logic
# Different discount rules based on customer tier AND order value

customer_tier = "Premium"
order_value = 3000

# First level: check customer tier
if customer_tier == "Premium":
    print("Premium Customer")
    
    # Second level: check order value (only for Premium customers)
    if order_value > 5000:
        # Premium customer with high-value order
        discount = 0.20
        print("High-value order: 20% discount")
    else:
        # Premium customer with standard order
        discount = 0.15
        print("Standard order: 15% discount")
else:
    # Non-premium customer - simple 5% discount regardless of order value
    discount = 0.05
    print("Standard customer: 5% discount")

print(f"Final discount: {discount:.0%}")
````
