## 1. Variables and Assignment

Variables are containers for storing data. Think of them as labeled boxes that hold information.

### Demo: Creating Variables

````python
# Demo 1: Simple variable assignment
# Variables store data with meaningful names - like labeled boxes
# Syntax: variable_name = value

# Create a variable to store a company code (text/string)
company_code = "TP01"

# Create a variable to store revenue (number)
revenue = 125000

# Create a variable to store region name (text/string)
region = "EMEA"

# print() displays the value stored in a variable
print(company_code)
print(revenue)
print(region)
````

````python
# Demo 2: Multiple assignment on one line
# You can assign multiple variables at once using commas
# Syntax: var1, var2 = value1, value2

# Assign two variables in a single line
product_id, product_name = "P-ENT", "Enterprise"

# Use f-strings to embed variables in text output
# f-string syntax: f"text {variable} more text"
print(f"Product ID: {product_id}")
print(f"Product Name: {product_name}")
````

````python
# Demo 3: Naming conventions (best practices)
# Good variable names are descriptive and use snake_case (lowercase with underscores)

# GOOD names - clear what each variable represents
total_revenue = 50000      # Total revenue in euros
cost_center = "MKT01"      # Cost center code
employee_count = 150       # Number of employees

# BAD names - avoid these patterns:
# x = 50000              # What is x? Not descriptive
# a = "MKT01"            # Single letter, meaningless
# TotalRevenue = 150     # Wrong case - use snake_case, not CamelCase

# Format numbers with commas using :, in f-strings
# Syntax: {variable:,} adds thousand separators
print(f"Total Revenue: €{total_revenue:,}")
print(f"Cost Center: {cost_center}")
print(f"Employees: {employee_count}")
````

### Variable Naming Rules:

**Rules (must follow):**
- Start with letter or underscore (not a number)
- Only letters, numbers, and underscores
- Case-sensitive (`revenue` ≠ `Revenue`)
- Can't use Python keywords (`if`, `for`, `True`, etc.)

**Conventions (best practices):**
- Use lowercase with underscores: `monthly_revenue`
- Be descriptive: `total_sales` not `ts`
- Use meaningful names: `customer_count` not `x`

## 2. Numeric Data Types

Python has two main numeric types:
- **int**: Whole numbers (e.g., 42, 1000, -15)
- **float**: Decimal numbers (e.g., 3.14, 99.99, -0.5)

### Demo: Working with Numbers

````python
# Demo 1: Integer (int) examples
# Integers are whole numbers - no decimal point
# Use for: counts, IDs, quantities

units = 42                 # Number of units (whole number)
employee_count = 150       # Count of employees
order_number = 1007768     # Order ID number

print(f"Units: {units}")
print(f"Employees: {employee_count}")
print(f"Order Number: {order_number}")
````

````python
# Demo 2: Float (decimal) examples
# Floats are numbers with decimal points
# Use for: prices, rates, percentages, precise measurements

revenue = 3082.59          # Revenue amount with cents
discount_rate = 0.15       # 15% as a decimal (15/100)
exchange_rate = 1.08       # Currency exchange rate

print(f"Revenue: €{revenue}")
print(f"Discount Rate: {discount_rate}")
print(f"Exchange Rate: {exchange_rate}")
````

````python
# Demo 3: Basic arithmetic operations
# Python supports all standard math operations

a = 10
b = 3

# Basic operations: + - * /
print(f"Addition: {a} + {b} = {a + b}")           # Add two numbers
print(f"Subtraction: {a} - {b} = {a - b}")        # Subtract
print(f"Multiplication: {a} * {b} = {a * b}")    # Multiply
print(f"Division: {a} / {b} = {a / b}")          # Divide (always returns float)
print(f"Exponentiation: {a} ** {b} = {a ** b}")  # Power: 10^3 = 1000

# Two more operators you'll use often:
# // Integer division - divides and rounds DOWN to whole number
print(f"\nInteger Division: {a} // {b} = {a // b}")

# % Modulo - returns the REMAINDER after division
print(f"Modulo (remainder): {a} % {b} = {a % b}")  # 10 ÷ 3 = 3 remainder 1
````

````python
# Demo 4: Practical use of integer division (//) and modulo (%)
# These operators are useful for packaging, scheduling, and validation

total_items = 47
box_capacity = 12

# How many FULL boxes can we fill?
# Use integer division // to get whole number result
full_boxes = total_items // box_capacity
print(f"Full boxes: {full_boxes}")  # 47 ÷ 12 = 3 full boxes

# How many items are LEFT OVER after filling full boxes?
# Use modulo % to get the remainder
leftover = total_items % box_capacity
print(f"Leftover items: {leftover}")  # 47 mod 12 = 11 leftover

# Common pattern: Check if a number is even or odd
# Even numbers have no remainder when divided by 2 (% 2 == 0)
order_number = 1007
if order_number % 2 == 0:
    print(f"{order_number} is even")
else:
    print(f"{order_number} is odd")
````

````python
# Demo 5: Business calculations
# Calculate selling price after discount

list_price = 499.00      # Original price
discount = 0.10          # 10% discount as decimal

# Calculate selling price: multiply by (1 - discount)
# This is: price × 0.90 = price × (1 - 0.10)
selling_price = list_price * (1 - discount)

print(f"List Price: €{list_price}")
print(f"Discount: {discount * 100}%")     # Convert decimal to percentage for display
print(f"Selling Price: €{selling_price}")
````

## 3. Strings

Strings are text data. Use quotes (single or double) to create them.

### Demo: Working with Strings

````python
# Demo 1: Creating strings
# Strings are text data - sequences of characters
# Create using single quotes '' or double quotes "" (both work the same)

product_name = "Enterprise"    # Using double quotes
category = "Suite"             # Using double quotes
status = 'Active'              # Using single quotes - same result

print(product_name)
print(category)
print(status)
````

````python
# Demo 2: String concatenation (combining strings)
# The old way: use + to join strings together

first_name = "John"
last_name = "Smith"

# Concatenate with + (must include space manually)
full_name = first_name + " " + last_name
print(full_name)

# Note: This works but f-strings (next demo) are cleaner
````

````python
# Demo 3: F-strings (modern, preferred way)
# F-strings let you embed variables directly inside strings
# Syntax: f"text {variable} more text"
# The f before the quote marks it as a formatted string

product = "Enterprise"
price = 499.00
category = "Suite"

# Create a formatted description with embedded variables
# Variables inside {} are automatically converted to strings
description = f"{product} - {category} (€{price})"
print(description)

# You can also do calculations inside the curly braces
message = f"Price with 10% discount: €{price * 0.9}"
print(message)
````

````python
# Demo 4: Formatting numbers in f-strings
# Use format specifiers after : inside the curly braces

revenue = 4987.88
large_number = 1250000

# ,.2f = commas for thousands + 2 decimal places
# Breakdown: , (comma separator) + .2 (2 decimals) + f (float format)
print(f"Revenue: €{revenue:,.2f}")

# :, = just commas, no decimal formatting
print(f"Large Number: €{large_number:,}")

# Format as percentage
# .1% = 1 decimal place + convert to percentage (multiplies by 100)
margin = 0.2534
print(f"Margin: {margin:.1%}")  # Shows as 25.3%
````

### String Formatting Reference:

```python
value = 1234.5678

f"{value}"           # 1234.5678
f"{value:.2f}"       # 1234.57 (2 decimal places)
f"{value:,.2f}"      # 1,234.57 (commas + 2 decimals)
f"{value:10.2f}"     # "   1234.57" (10 chars wide)

percent = 0.156
f"{percent:.1%}"     # 15.6%
```

## 4. Booleans

Booleans represent True or False values. Essential for decision-making in code.

### Demo: Boolean Values and Comparisons

````python
# Demo 1: Boolean values
# Booleans can only be True or False (note the capital letters)
# Use for: flags, status indicators, yes/no conditions

is_active = True           # Account is active
has_discount = False       # No discount applied
in_stock = True            # Item is in stock

print(f"Active: {is_active}")
print(f"Has Discount: {has_discount}")
print(f"In Stock: {in_stock}")
````

````python
# Demo 2: Comparison operators create booleans
# Comparisons evaluate to True or False

revenue = 5000
threshold = 3000

# > greater than: is revenue larger than threshold?
exceeds_threshold = revenue > threshold  # True (5000 > 3000)

# >= greater than or equal: is revenue at least 1000?
meets_minimum = revenue >= 1000  # True (5000 >= 1000)

# == equals: is revenue exactly zero?
is_zero = revenue == 0  # False (5000 is not 0)

print(f"Exceeds threshold: {exceeds_threshold}")
print(f"Meets minimum: {meets_minimum}")
print(f"Is zero: {is_zero}")
````

````python
# Demo 3: All comparison operators
# These all return True or False

a = 10
b = 5

# == Equal to (note: double equals for comparison, single for assignment)
print(f"a == b: {a == b}")  # False - 10 is not equal to 5

# != Not equal to
print(f"a != b: {a != b}")  # True - 10 is not equal to 5

# > Greater than
print(f"a > b: {a > b}")    # True - 10 is greater than 5

# < Less than
print(f"a < b: {a < b}")    # False - 10 is not less than 5

# >= Greater than or equal
print(f"a >= b: {a >= b}")  # True - 10 is greater than or equal to 5

# <= Less than or equal
print(f"a <= b: {a <= b}")  # False - 10 is not less than or equal to 5
````

## 5. None - Representing "Nothing"

Python has a special value `None` that represents "no value" or "unknown". You'll encounter it when data is missing.

````python
# Demo: Working with None
# None is Python's way of saying "no value" or "nothing here"
# It's different from 0, empty string, or False - it means absence of any value

# Common use cases for None:
# - Data hasn't been provided yet
# - A value is missing or unknown
# - Explicitly clearing a value

# Example: Customer hasn't entered a discount code yet
discount_code = None

# Example: No salesperson has been assigned to this account
assigned_salesperson = None

print(f"Discount code: {discount_code}")

# type() shows that None has its own special type: NoneType
print(f"Type of None: {type(discount_code)}")

# Checking if a value is None
# Use "is None" (not "== None") - this is the Pythonic way
if discount_code is None:
    print("No discount code entered")

# Important: None is NOT the same as other "empty" values
empty_string = ""   # This is a string with zero characters
zero_value = 0      # This is the number zero
no_value = None     # This is "no value at all"

# Use "is None" to test - all these return False except the actual None
print(f"\nEmpty string is None: {empty_string is None}")  # False - it's a string
print(f"Zero is None: {zero_value is None}")              # False - it's an int
print(f"None is None: {no_value is None}")                # True - it IS None

# Common pattern: Provide a default value when None
# Syntax: value if condition else default_value
# This says: "use discount_code if it's not None, otherwise use 0.0"
discount_rate = discount_code if discount_code is not None else 0.0
print(f"\nDiscount rate (with default): {discount_rate}")
````

## 5. Type Checking and Conversion

Python has a `type()` function to check data types, and functions to convert between types.

### Demo: Checking and Converting Types

````python
# Demo 1: Checking types with type()
# The type() function tells you what data type a value is
# Useful for debugging and validating data

revenue = 5000           # An integer (whole number)
price = 99.99            # A float (decimal number)
product = "Enterprise"   # A string (text)
is_active = True         # A boolean (True/False)

# type() returns the class of the value
# <class 'int'> means it's an integer
# <class 'float'> means it's a float
# <class 'str'> means it's a string
# <class 'bool'> means it's a boolean
print(f"Type of {revenue}: {type(revenue)}")
print(f"Type of {price}: {type(price)}")
print(f"Type of '{product}': {type(product)}")
print(f"Type of {is_active}: {type(is_active)}")
````

````python
# Demo 2: Converting between types
# Use conversion functions: int(), float(), str()
# These create a NEW value of the target type

# String to integer: int()
# Useful when reading numeric data from files (CSV data is all strings)
order_id_str = "1007768"              # This is a string (text)
order_id_int = int(order_id_str)      # Convert to integer
print(f"Converted '{order_id_str}' to {order_id_int}")
print(f"Type: {type(order_id_int)}")

# String to float: float()
# Use for decimal numbers that come as strings
price_str = "499.00"                  # This is a string
price_float = float(price_str)        # Convert to float
print(f"\nConverted '{price_str}' to {price_float}")
print(f"Type: {type(price_float)}")

# Number to string: str()
# Use when you need to combine numbers with text, or write to files
revenue = 5000                        # This is an integer
revenue_str = str(revenue)            # Convert to string
print(f"\nConverted {revenue} to '{revenue_str}'")
print(f"Type: {type(revenue_str)}")
````

````python
# Demo 3: Why type conversion matters - a common scenario
# When you read data from CSV files, ALL values come as strings!
# You must convert them to do math

# Simulating data from a CSV file (everything is text)
units_str = "25"
price_str = "199.00"

# PROBLEM: Can't do math with strings!
# Uncommenting the line below would cause an error:
# total = units_str * price_str  # TypeError! Can't multiply strings this way

# SOLUTION: Convert to numbers first, then calculate
# Use int() for whole numbers
units = int(units_str)

# Use float() for decimal numbers  
price = float(price_str)

# Now we can do math!
total = units * price

# Show the results with their types to confirm conversion worked
print(f"Units: {units} (type: {type(units).__name__})")     # int
print(f"Price: {price} (type: {type(price).__name__})")     # float
print(f"Total: €{total:,.2f}")  # 25 × 199.00 = 4,975.00
````
