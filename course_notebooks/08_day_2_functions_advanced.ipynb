## 2. Default Parameters

Default parameters let you define **optional** values that are used if the caller doesn't provide them.

### Why Use Default Parameters?
- Make functions easier to use for common cases
- Reduce the need to always specify every argument
- Allow functions to grow without breaking existing code

### Demo 2.1: Basic Default Parameters

````python
# Demo: Default Parameters
# Assign a default value with = in the parameter list
# Parameters with defaults become OPTIONAL

# Without default: Must ALWAYS provide tax rate
def calculate_tax_v1(amount, rate):
    return amount * rate

# calculate_tax_v1(1000)  # Error! Missing required argument: rate

# With default: Rate is optional (defaults to 20% if not specified)
# Syntax: parameter_name=default_value
def calculate_tax(amount, rate=0.20):
    """Calculate tax. Rate defaults to 20% if not specified."""
    return amount * rate

# Use the default rate - don't specify rate at all
tax1 = calculate_tax(1000)
print(f"Tax at default rate: €{tax1:.2f}")

# Override with a different rate - provide your own value
tax2 = calculate_tax(1000, 0.25)
print(f"Tax at 25%: €{tax2:.2f}")

tax3 = calculate_tax(1000, 0.07)
print(f"Tax at 7%: €{tax3:.2f}")
````

### Demo 2.2: Multiple Default Parameters

````python
# Demo: Multiple Default Parameters
# You can have several parameters with defaults
# Required parameters must come BEFORE default parameters

def format_currency(amount, currency="€", decimals=2, show_symbol=True):
    """
    Format a number as currency.
    
    Parameters:
    amount: The number to format (required)
    currency: Currency symbol (default: €)
    decimals: Decimal places (default: 2)
    show_symbol: Whether to show currency symbol (default: True)
    """
    if show_symbol:
        # Format with symbol - {decimals} uses the variable value
        return f"{currency}{amount:,.{decimals}f}"
    else:
        # Format without symbol
        return f"{amount:,.{decimals}f}"

# Use all defaults - just provide the required 'amount'
print(format_currency(1500))                    # €1,500.00

# Override just the first default (currency)
print(format_currency(1500, "$"))               # $1,500.00

# Override first two defaults (currency and decimals)
print(format_currency(1500, "£", 0))            # £1,500

# Override all parameters
print(format_currency(1500, "$", 2, False))     # 1,500.00
````

### Demo 2.3: Business Example - Discount Calculator

````python
# Demo: Business Example - Discount Calculator with defaults

def apply_discount(price, discount_rate=0.10, min_price=0):
    """
    Apply discount to a price.
    
    Parameters:
    price: Original price (required)
    discount_rate: Discount as decimal (default: 10%)
    min_price: Minimum price floor (default: 0)
    
    Returns:
    Discounted price (not less than min_price)
    """
    # Calculate discounted price: price × (1 - discount)
    # Example: €100 × (1 - 0.10) = €100 × 0.90 = €90
    discounted = price * (1 - discount_rate)
    
    # max() returns the larger of two values
    # This ensures we don't go below min_price
    return max(discounted, min_price)

# Standard 10% discount (using default rate)
print(f"€199 with default discount: €{apply_discount(199):.2f}")

# Premium 25% discount (overriding rate)
print(f"€199 with 25% discount: €{apply_discount(199, 0.25):.2f}")

# Discount with minimum price floor (prevents going too low)
# €50 × 0.20 = €10, but min is €20, so result is €20
print(f"€50 with 80% discount (min €20): €{apply_discount(50, 0.80, 20):.2f}")
````

## 3. Keyword Arguments

When calling functions, you can specify arguments **by name**. This makes code more readable and allows you to skip optional parameters.

### Demo 3.1: Positional vs Keyword Arguments

````python
# Demo: Positional vs Keyword Arguments
# Keyword arguments use parameter_name=value syntax

def create_order(product, quantity, price, discount=0, priority=False):
    """
    Create an order summary.
    """
    subtotal = quantity * price
    total = subtotal * (1 - discount)
    status = "PRIORITY" if priority else "Standard"
    return f"{product} x{quantity}: €{total:.2f} [{status}]"

# Positional arguments: VALUES only, ORDER matters!
# Must match the parameter order exactly
order1 = create_order("Suite A", 5, 199, 0.10, True)
print(f"Positional: {order1}")

# Keyword arguments: NAME=value pairs, ORDER doesn't matter
# Much clearer what each value means
order2 = create_order(
    product="Suite A",
    quantity=5,
    price=199,
    discount=0.10,
    priority=True
)
print(f"Keyword: {order2}")

# Mixed: positional first, then keyword
# Can skip optional parameters and specify by name
order3 = create_order("Suite B", 3, 99, priority=True)  # Skip discount!
print(f"Mixed: {order3}")
````

### Demo 3.2: Skipping Optional Parameters with Keywords

````python
# Demo: Skipping Optional Parameters with Keywords
# Keyword arguments let you use defaults for some params
# while specifying values for others out of order

def generate_report(title, data, show_header=True, show_total=True, 
                    currency="€", decimal_places=2):
    """
    Generate a formatted report.
    """
    lines = []
    
    if show_header:
        lines.append(f"=== {title} ===")
    
    for item, value in data.items():
        lines.append(f"{item}: {currency}{value:,.{decimal_places}f}")
    
    if show_total:
        total = sum(data.values())
        lines.append(f"TOTAL: {currency}{total:,.{decimal_places}f}")
    
    return "\n".join(lines)

# Sample data (dictionary)
sales = {"EMEA": 125000, "AMER": 98000, "APAC": 67000}

# Default report - use all defaults
print(generate_report("Q1 Sales", sales))
print()

# Skip to specific parameters using keywords
# Uses defaults for show_header and show_total
# Overrides currency and decimal_places
print(generate_report("Q1 Sales", sales, currency="$", decimal_places=0))
````

### Best Practice: When to Use Keywords

```python
# BAD: What do these values mean?
result = process_data(data, True, False, 100, "high")

# GOOD: Clear what each parameter is
result = process_data(
    data,
    include_nulls=True,
    sort_ascending=False,
    limit=100,
    priority="high"
)
```

**Rule of thumb**: Use keywords for:
- Boolean parameters
- Functions with 3+ parameters
- When skipping optional parameters

## 4. Functions with Lists

Functions become very powerful when they process lists of data. There are common patterns you'll use repeatedly.

### Demo 4.1: Pattern 1 - Transform (Map)

````python
# Demo: Pattern 1 - Transform (Map)
# Apply the same operation to every item in a list
# Returns a NEW list with transformed values

def apply_discount_to_all(prices, discount_rate):
    """
    Apply discount to every price in the list.
    Returns a NEW list with discounted prices.
    """
    # Start with an empty list to collect results
    discounted = []
    
    # Process each price one by one
    for price in prices:
        # Apply the discount calculation
        new_price = price * (1 - discount_rate)
        # Add to results
        discounted.append(new_price)
    
    # Return the new list
    return discounted

# Test it
original_prices = [199, 99, 299, 499]
sale_prices = apply_discount_to_all(original_prices, 0.15)

print(f"Original: {original_prices}")
print(f"With 15% off: {[f'€{p:.2f}' for p in sale_prices]}")
````

### Demo 4.2: Pattern 2 - Filter

````python
# Demo: Pattern 2 - Filter
# Keep only items that meet a condition
# Returns a NEW list with only qualifying items

def filter_large_transactions(amounts, threshold):
    """
    Keep only transactions above threshold.
    Returns a NEW list with filtered items.
    """
    # Start with empty list for qualifying items
    large = []
    
    # Check each amount
    for amount in amounts:
        # Only include if above threshold
        if amount > threshold:
            large.append(amount)
    
    return large

# Test it
all_transactions = [250, 1500, 75, 3200, 450, 8000, 125]
large_only = filter_large_transactions(all_transactions, 1000)

print(f"All transactions: {all_transactions}")
print(f"Large (>€1000): {large_only}")
````

### Demo 4.3: Pattern 3 - Reduce (Aggregate)

````python
# Demo: Pattern 3 - Reduce (Aggregate)
# Combine all items into a single result (or summary)
# Returns summary statistics instead of a list

def calculate_statistics(values):
    """
    Calculate multiple statistics from a list.
    Returns a dictionary with all statistics.
    """
    # Handle edge case: empty list
    if not values:
        return {"count": 0, "total": 0, "average": 0, "min": None, "max": None}
    
    # Use built-in functions to calculate statistics
    return {
        "count": len(values),               # Number of items
        "total": sum(values),               # Sum of all values
        "average": sum(values) / len(values),  # Mean
        "min": min(values),                 # Smallest value
        "max": max(values)                  # Largest value
    }

# Test it
revenues = [3082.59, 556.34, 2422.68, 4155.00, 1876.50]
stats = calculate_statistics(revenues)

print("Revenue Statistics:")
print(f"  Count: {stats['count']}")
print(f"  Total: €{stats['total']:,.2f}")
print(f"  Average: €{stats['average']:,.2f}")
print(f"  Range: €{stats['min']:,.2f} - €{stats['max']:,.2f}")
````

### Demo 4.4: Combining Patterns

````python
# Demo: Combining Patterns - Filter → Transform → Aggregate
# Real analysis often chains multiple patterns together

def analyze_sales(amounts, min_threshold=0, apply_discount=0):
    """
    Complete sales analysis: filter, transform, and aggregate.
    
    Pipeline:
    1. Filter: Keep only amounts above min_threshold
    2. Transform: Apply discount if specified
    3. Aggregate: Calculate statistics
    """
    # Step 1: Filter - keep only amounts >= threshold
    # List comprehension: [item for item in list if condition]
    filtered = [amt for amt in amounts if amt >= min_threshold]
    
    # Step 2: Transform - apply discount to each
    if apply_discount > 0:
        processed = [amt * (1 - apply_discount) for amt in filtered]
    else:
        processed = filtered  # No discount, use filtered as-is
    
    # Step 3: Aggregate - calculate summary statistics
    return {
        "original_count": len(amounts),
        "filtered_count": len(filtered),
        "total": sum(processed),
        "average": sum(processed) / len(processed) if processed else 0
    }

# Test it with chained operations
sales = [150, 2500, 75, 1800, 3200, 450, 5000]

result = analyze_sales(sales, min_threshold=500, apply_discount=0.10)

print(f"Original transactions: {result['original_count']}")
print(f"After filtering (>=€500): {result['filtered_count']}")
print(f"Total (after 10% discount): €{result['total']:,.2f}")
print(f"Average: €{result['average']:,.2f}")
````

## 5. In-Place Modification vs Returning New Lists

**Important concept**: Functions can either modify the original list OR return a new list. Understanding the difference prevents bugs!

### Demo 5.1: The Difference

````python
# Demo: In-Place Modification vs Returning New Lists
# CRITICAL concept: Lists can be modified in place or copied

# VERSION 1: Modifies the ORIGINAL list (in-place)
def add_tax_inplace(prices, tax_rate=0.20):
    """
    Adds tax to prices IN PLACE.
    WARNING: This changes the original list!
    """
    # Use index to modify each element of the original list
    for i in range(len(prices)):
        prices[i] = prices[i] * (1 + tax_rate)
    # No return needed - original is modified directly

# VERSION 2: Returns a NEW list (original unchanged)
def add_tax_new(prices, tax_rate=0.20):
    """
    Returns a NEW list with tax added.
    Original list is NOT changed.
    """
    # List comprehension creates a brand new list
    return [price * (1 + tax_rate) for price in prices]

# Demonstrate the DIFFERENCE
print("=== In-Place Modification ===")
prices1 = [100, 200, 300]
print(f"Before: {prices1}")
add_tax_inplace(prices1)      # Modifies prices1 directly
print(f"After:  {prices1}")   # Original is CHANGED!

print("\n=== Returning New List ===")
prices2 = [100, 200, 300]
print(f"Before: {prices2}")
with_tax = add_tax_new(prices2)  # Creates new list
print(f"After:  {prices2}")      # Original UNCHANGED!
print(f"New:    {with_tax}")     # New list has changes
````

### Best Practice Recommendation

**Usually prefer returning new lists:**
- Easier to debug (original data preserved)
- Avoids unexpected side effects
- More "functional" programming style
- Can chain operations together

**Use in-place when:**
- Working with very large lists (memory efficiency)
- The modification is the primary purpose
- You explicitly want to change the original

### Demo 5.2: Common Mistake to Avoid

````python
# Demo: Common Mistake - Accidentally modifying input list
# Lists are passed by REFERENCE, not by value

def process_data(data):
    """This function MODIFIES the input list by accident!"""
    data.append(999)  # This changes the original!
    data.sort()       # This also changes the original!
    return data

original = [5, 2, 8, 1]
print(f"Original before: {original}")

result = process_data(original)

# SURPRISE! The original list is changed!
print(f"Original after:  {original}")  # Now [1, 2, 5, 8, 999]
print(f"Result:          {result}")
# They're actually the SAME object in memory
print(f"Same object?     {original is result}")  # True!
````

````python
# Demo: Safe Version - Make a copy first
# Use .copy() to create an independent copy

def process_data_safe(data):
    """This function returns a NEW list, original unchanged."""
    # Create a copy FIRST before making any changes
    result = data.copy()  # Independent copy
    result.append(999)     # Modify the copy
    result.sort()          # Sort the copy
    return result

original = [5, 2, 8, 1]
print(f"Original before: {original}")

result = process_data_safe(original)

# Original is SAFE - unchanged
print(f"Original after:  {original}")  # Still [5, 2, 8, 1]
print(f"Result:          {result}")     # New sorted list with 999
print(f"Same object?     {original is result}")  # False - different lists!
````

## 6. Practical Example: Sales Data Processor

Let's build a complete, practical function that combines all concepts:

````python
def process_sales_data(transactions, 
                       min_amount=0, 
                       max_amount=float('inf'),
                       apply_discount=0,
                       include_stats=True):
    """
    Process sales transaction data with filtering and optional discount.
    
    Parameters:
    transactions: List of transaction amounts
    min_amount: Minimum amount to include (default: 0)
    max_amount: Maximum amount to include (default: no limit)
    apply_discount: Discount rate to apply (default: 0)
    include_stats: Whether to calculate statistics (default: True)
    
    Returns:
    Dictionary with processed data and optional statistics
    """
    # Step 1: Filter transactions
    filtered = [t for t in transactions if min_amount <= t <= max_amount]
    
    # Step 2: Apply discount if specified
    if apply_discount > 0:
        processed = [t * (1 - apply_discount) for t in filtered]
    else:
        processed = filtered.copy()  # Always return a new list
    
    # Build result
    result = {
        "original_count": len(transactions),
        "filtered_count": len(filtered),
        "processed_transactions": processed
    }
    
    # Step 3: Add statistics if requested
    if include_stats and processed:
        result["statistics"] = {
            "total": sum(processed),
            "average": sum(processed) / len(processed),
            "min": min(processed),
            "max": max(processed)
        }
    
    return result

# Example usage
daily_sales = [150, 2500, 75, 1800, 3200, 450, 5000, 890, 1250]

# Process with filters and discount
result = process_sales_data(
    daily_sales,
    min_amount=500,
    max_amount=4000,
    apply_discount=0.10,
    include_stats=True
)

print("=" * 50)
print("SALES PROCESSING REPORT")
print("=" * 50)
print(f"Original transactions: {result['original_count']}")
print(f"After filtering (€500-€4000): {result['filtered_count']}")
print(f"\nProcessed amounts (with 10% discount):")
for amount in result['processed_transactions']:
    print(f"  €{amount:,.2f}")
print(f"\nStatistics:")
print(f"  Total: €{result['statistics']['total']:,.2f}")
print(f"  Average: €{result['statistics']['average']:,.2f}")
print(f"  Range: €{result['statistics']['min']:,.2f} - €{result['statistics']['max']:,.2f}")
print("=" * 50)
````

## Challenge Exercise: Transaction Analyzer

Build a complete transaction analysis function that:

1. Takes a list of transaction dictionaries: `[{"id": str, "amount": float, "category": str}, ...]`
2. Has optional parameters:
   - `categories` (list): Filter to only these categories (default: all)
   - `min_amount` (float): Minimum amount (default: 0)
   - `sort_by` (str): "amount" or "id" (default: "amount")
   - `ascending` (bool): Sort order (default: False = descending)

3. Returns a dictionary with:
   - `transactions`: Filtered and sorted list
   - `total`: Sum of amounts
   - `count`: Number of transactions
   - `by_category`: Dictionary of totals per category

````python
def analyze_transactions(transactions, categories=None, min_amount=0,
                         sort_by="amount", ascending=False):
    """
    Analyze transactions with filtering and sorting.
    
    Parameters:
    transactions: List of transaction dicts with 'id', 'amount', 'category'
    categories: List of categories to include (None = all)
    min_amount: Minimum transaction amount
    sort_by: Field to sort by ('amount' or 'id')
    ascending: Sort order
    
    Returns:
    Dictionary with analysis results
    """
    # Step 1: Filter by category (if specified)
    if categories:
        filtered = [t for t in transactions if t["category"] in categories]
    else:
        filtered = transactions.copy()
    
    # Step 2: Filter by minimum amount
    filtered = [t for t in filtered if t["amount"] >= min_amount]
    
    # Step 3: Sort
    # Hint: sorted(list, key=lambda x: x["field"], reverse=True/False)
    sorted_transactions = sorted(
        filtered,
        key=lambda x: x[sort_by],
        reverse=not ascending
    )
    
    # Step 4: Calculate totals by category
    by_category = {}
    for t in sorted_transactions:
        cat = t["category"]
        if cat not in by_category:
            by_category[cat] = 0
        by_category[cat] += t["amount"]
    
    # Step 5: Build result
    return {
        "transactions": sorted_transactions,
        "count": len(sorted_transactions),
        "total": sum(t["amount"] for t in sorted_transactions),
        "by_category": by_category
    }

# Test data
test_transactions = [
    {"id": "TXN-001", "amount": 1500, "category": "Software"},
    {"id": "TXN-002", "amount": 250, "category": "Services"},
    {"id": "TXN-003", "amount": 3200, "category": "Software"},
    {"id": "TXN-004", "amount": 800, "category": "Hardware"},
    {"id": "TXN-005", "amount": 5000, "category": "Software"},
    {"id": "TXN-006", "amount": 450, "category": "Services"},
]

# Test: Software transactions over €1000, sorted by amount descending
result = analyze_transactions(
    test_transactions,
    categories=["Software"],
    min_amount=1000,
    sort_by="amount",
    ascending=False
)

print("Analysis Results:")
print(f"Count: {result['count']}")
print(f"Total: €{result['total']:,.2f}")
print(f"\nTransactions:")
for t in result['transactions']:
    print(f"  {t['id']}: €{t['amount']:,.2f} ({t['category']})")
print(f"\nBy Category: {result['by_category']}")
````
