## 1. Function Basics

Functions are reusable blocks of code that perform specific tasks. They help you:
- **Avoid repeating code** - Write once, use many times
- **Make code more readable** - Named functions explain what code does
- **Easier to test and debug** - Isolate problems to specific functions

**Syntax:**
```python
def function_name():
    # code to execute
    pass
```

### Demo: Simple Functions

````python
# Demo 1: Simple function without parameters
# A function is defined using the 'def' keyword followed by a name and parentheses
# The function body is indented below the definition

# Syntax: def function_name():
#             code to execute

def welcome_message():
    # This code runs when the function is called
    print("Welcome to Data Analytics!")
    print("Let's analyze some business data.")

# To run a function, "call" it by writing its name followed by ()
# Nothing happens until you call the function!
welcome_message()
````

````python
# Demo 2: Function that performs a calculation
# Functions can contain any code - variables, calculations, etc.

def calculate_vat():
    # Variables created inside a function are "local" to that function
    amount = 1000
    vat_rate = 0.20
    vat = amount * vat_rate
    print(f"VAT on €{amount} is €{vat}")

# Call the function to run the code
calculate_vat()
````

````python
# Demo 3: Calling functions multiple times
# The main benefit of functions: define once, use many times!

def print_separator():
    # This simple function prints a line of 50 equals signs
    print("=" * 50)

# We can call the same function as many times as needed
# Each call runs the function's code
print("Monthly Sales Report")
print_separator()  # First call
print("Region: EMEA")
print("Total Revenue: €125,000")
print_separator()  # Second call - reuses the same code
````

## 2. Parameters and Return Values

Functions become much more powerful when they can:
- Accept **parameters** (input values)
- **Return** results (output values)

This makes functions flexible and reusable with different data.

### Demo: Functions with Parameters

````python
# Demo 1: Function with one parameter
# Parameters are inputs - variables that receive values when function is called
# Put parameter names inside the parentheses

def greet_employee(name):
    # 'name' is a parameter - it will hold whatever value is passed in
    print(f"Hello, {name}! Welcome to the analytics team.")

# When calling, provide an "argument" (actual value) for each parameter
# The argument value gets assigned to the parameter variable
greet_employee("Anna")    # name = "Anna"
greet_employee("Marcus")  # name = "Marcus"
greet_employee("Sophie")  # name = "Sophie"
````

````python
# Demo 2: Function with multiple parameters
# Separate multiple parameters with commas
# Order matters: arguments match parameters by position

def calculate_revenue(units_sold, price_per_unit):
    # units_sold = first argument, price_per_unit = second argument
    revenue = units_sold * price_per_unit
    print(f"Revenue: €{revenue:,.2f}")

# Call with different combinations of values
# First argument → units_sold, Second argument → price_per_unit
calculate_revenue(150, 49.99)     # 150 units × €49.99
calculate_revenue(75, 199.00)     # 75 units × €199.00
calculate_revenue(1200, 5.50)     # 1200 units × €5.50
````

````python
# Demo 3: Function with return value
# 'return' sends a value back to where the function was called
# The returned value can be stored in a variable or used in expressions

def calculate_margin(revenue, cost):
    # Perform calculations
    profit = revenue - cost
    margin_percent = (profit / revenue) * 100
    
    # Return the result - this value is sent back to the caller
    return margin_percent

# Store the returned value in a variable
# The function call "becomes" the returned value
product_a_margin = calculate_margin(5000, 3200)  # Returns 36.0
product_b_margin = calculate_margin(8000, 5600)  # Returns 30.0

print(f"Product A Margin: {product_a_margin:.1f}%")
print(f"Product B Margin: {product_b_margin:.1f}%")

# We can use returned values directly in calculations
average_margin = (product_a_margin + product_b_margin) / 2
print(f"Average Margin: {average_margin:.1f}%")
````

## 3. Returning Multiple Values

Python functions can return multiple values at once using tuples.
This is very useful for business calculations that produce several related results.

### Demo: Multiple Return Values

````python
# Demo 1: Returning multiple values
# Python allows returning multiple values separated by commas
# This actually returns a tuple that can be unpacked into variables

def analyze_transaction(amount, tax_rate):
    # Calculate two values
    tax = amount * tax_rate
    total = amount + tax
    
    # Return both values, separated by comma
    # Python packages them as a tuple automatically
    return tax, total

# Unpack the returned tuple into separate variables
# Number of variables must match number of returned values
transaction_tax, transaction_total = analyze_transaction(1000, 0.20)

print(f"Tax: €{transaction_tax:.2f}")
print(f"Total: €{transaction_total:.2f}")
````

````python
# Demo 2: Complete sales analysis function returning three values
# Useful when calculations produce multiple related results

def calculate_sales_metrics(revenue, cost):
    # Calculate profit: revenue minus cost
    profit = revenue - cost
    
    # Calculate margin: profit as percentage of revenue
    margin_percent = (profit / revenue) * 100
    
    # Calculate ROI: profit as percentage of cost (return on investment)
    roi = (profit / cost) * 100
    
    # Return all three values
    return profit, margin_percent, roi

# Unpack all three values into separate variables
# Variable names should match what each value represents
profit, margin, roi = calculate_sales_metrics(15000, 9000)

print(f"Profit: €{profit:,.2f}")
print(f"Margin: {margin:.1f}%")
print(f"ROI: {roi:.1f}%")
````

## 4. Documenting Functions with Docstrings

**Docstrings** are special comments that describe what a function does.
They make your code easier to understand and maintain.

Use triple quotes `"""` right after the function definition.

### Demo: Docstrings

````python
# Demo 1: Basic docstring
# A docstring is a string that documents what a function does
# Place it immediately after the def line, inside triple quotes

def calculate_vat(amount):
    """
    Calculate VAT at 20% on a given amount.
    """
    return amount * 0.20

# You can access the docstring using __doc__ attribute
# This is useful for understanding functions you didn't write
print(calculate_vat.__doc__)

# The function still works normally
print(f"VAT: €{calculate_vat(1000):.2f}")
````

````python
# Demo 2: Detailed docstring with parameters and return description
# Best practice: document what each parameter is and what the function returns

def calculate_commission(sales_amount, commission_rate):
    """
    Calculate sales commission based on total sales and commission rate.
    
    Parameters:
    sales_amount (float): Total sales value in euros
    commission_rate (float): Commission rate as decimal (e.g., 0.05 for 5%)
    
    Returns:
    float: Commission amount in euros
    """
    return sales_amount * commission_rate

# Calculate commission for a sales representative
monthly_sales = 25000
commission = calculate_commission(monthly_sales, 0.08)

print(f"Sales: €{monthly_sales:,.2f}")
print(f"Commission (8%): €{commission:,.2f}")
````
