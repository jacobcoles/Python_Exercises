## 3. Essential Keyboard Shortcuts

### Running Cells:
- **Shift + Enter** - Run cell and move to next cell
- **Ctrl + Enter** (or Cmd + Enter on Mac) - Run cell and stay on same cell

### Editing Cells:
- **Enter** - Enter edit mode (start typing in cell)
- **Esc** - Exit edit mode (command mode)

### Adding/Deleting Cells (press Esc first):
- **A** - Insert cell Above
- **B** - Insert cell Below
- **DD** - Delete cell (press D twice)

### Cell Types (press Esc first):
- **Y** - Change to code cell
- **M** - Change to markdown cell

### Other Useful:
- **Z** - Undo delete cell
- **Shift + M** - Merge selected cells
- **Ctrl + S** (Cmd + S on Mac) - Save notebook

## 5. Understanding Code Output

### Demo: Different Types of Output

````python
# Output Demo 1: Automatic display of last expression
# Jupyter automatically shows the result of the LAST line in a cell
# You don't need print() for this - it just displays the value

5 + 3

# The result "8" appears below because it's the last expression
````

````python
# Output Demo 2: Only the LAST line displays automatically
# This is a common gotcha for beginners!

10 + 5     # This result (15) is NOT displayed
20 + 10    # Only this result (30) will display

# Problem: We lost the first calculation!
# Solution: Use print() when you need to see multiple values
````

````python
# Output Demo 3: Use print() to display multiple values
# Store results in variables, then print each one

# Calculate and store in variables
result1 = 10 + 5
result2 = 20 + 10

# Use print() with f-strings to display both results
# Each print() outputs on a new line
print(f"First calculation: {result1}")
print(f"Second calculation: {result2}")

# Now we can see both values!
````

## 6. Comments in Code

### Why Use Comments?
- Explain what your code does
- Leave notes for yourself and others
- Temporarily disable code

### Demo: Comment Syntax

````python
# Demo: Single-line comments
# This is a single-line comment - starts with #
# Python completely ignores everything after # on that line

# Use comments to explain WHAT your code does and WHY
revenue = 50000  # You can also add comments at end of lines

# Calculate 20% VAT
# Good comment: explains the business purpose
vat = revenue * 0.20

# Display results using f-strings with thousand separators (:,)
print(f"Revenue: €{revenue:,}")
print(f"VAT: €{vat:,}")
````

````python
# Demo: Multi-line comments (docstrings)
# Use triple quotes for longer explanations

"""
This is a multi-line comment (also called a docstring).
You can write several lines of explanation here.
Useful for:
- Documenting what a section of code does
- Explaining complex business logic
- Temporarily disabling blocks of code
"""

# The triple-quoted text above is ignored by Python
print("Comments don't affect code execution")
````

## 7. Best Practices for Jupyter Notebooks

### DO:
✓ Run cells in order from top to bottom
✓ Save frequently (Ctrl+S / Cmd+S)
✓ Use clear variable names
✓ Add comments to explain complex logic
✓ Use markdown cells to organize sections
✓ Test code in small chunks

### DON'T:
✗ Run cells out of order (can cause confusion)
✗ Use single letters as variable names (x, y, z)
✗ Write very long cells (break them up)
✗ Forget to save your work

### Restart and Run All:
Before finishing, test that your notebook works from start to finish:
- **Kernel** → **Restart & Run All**
- This ensures cells work in the correct order
