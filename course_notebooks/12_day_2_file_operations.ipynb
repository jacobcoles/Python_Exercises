## 1. Reading CSV FilesMost business data comes in CSV format.

```python
# Demo: Reading CSV Files with the csv Module
# CSV = Comma-Separated Values - the most common data file format
# Python's built-in csv module handles reading/writing CSV files

import csv

# Open and read a CSV file
# Syntax: with open("filename", "r") as file:
# The 'with' statement ensures the file is properly closed after use
# "r" = read mode

# csv.DictReader() reads each row as a dictionary
# Keys = column headers, Values = row data
with open("sales.csv", "r") as file:
    reader = csv.DictReader(file)  # Create a reader object
    
    # Loop through each row (each row is a dictionary)
    for row in reader:
        # Access values using column names as keys
        print(f"{row['Product']}: €{row['Revenue']}")
```

### Exercise: Process CSV data

```python
# Exercise: Process CSV Data - Calculate Total Revenue
# Task: Read a CSV file and sum up all revenue values

import csv

# Initialize a variable to accumulate the total
total_revenue = 0

# Open the file and create a DictReader
with open("sales.csv", "r") as file:
    reader = csv.DictReader(file)
    
    # Loop through each row
    for row in reader:
        # CSV values are always strings - convert to float for math
        # Use float() because revenue might have decimals
        total_revenue += float(row["Revenue"])

# Display the result with formatting
# :,.2f = comma thousands separator, 2 decimal places
print(f"Total: €{total_revenue:,.2f}")
```

## 2. Writing CSV FilesSave processed data back to files.

```python
# Demo: Writing CSV Files
# Save processed data back to a new CSV file

import csv

# Data to write - list of dictionaries (each dict = one row)
data = [
    {"Product": "Suite A", "Price": 199},
    {"Product": "Suite B", "Price": 99}
]

# Open file for writing
# "w" = write mode (creates new file or overwrites existing)
# newline="" prevents extra blank rows on Windows
with open("output.csv", "w", newline="") as file:
    # Create a DictWriter - specify the column names with fieldnames
    writer = csv.DictWriter(file, fieldnames=["Product", "Price"])
    
    # Write the header row (column names)
    writer.writeheader()
    
    # Write all data rows at once
    writer.writerows(data)

print("File written successfully")
```
