## 1. GroupBy BasicsGroup data by category and calculate statistics.

```python
# Demo: GroupBy Basics - The Split-Apply-Combine Pattern
# groupby() is ESSENTIAL for summarizing data by categories
# Pattern: Split data into groups → Apply a function → Combine results
# Syntax: df.groupby("column")["value_column"].aggregation()

import pandas as pd

df = pd.DataFrame({
    "Region": ["EMEA", "AMER", "EMEA", "APAC", "AMER"],
    "Revenue": [15000, 12000, 8500, 9500, 11000]
})

# Group by Region and sum the Revenue
# Step 1: groupby("Region") - splits data into EMEA, AMER, APAC groups
# Step 2: ["Revenue"] - select the column to aggregate
# Step 3: .sum() - apply the sum function to each group
regional_total = df.groupby("Region")["Revenue"].sum()
print(regional_total)
```

## 2. Multiple AggregationsCalculate several statistics at once.

```python
# Demo: Multiple Aggregations with .agg()
# Calculate several statistics at once using .agg()
# Syntax: df.groupby("column")["value"].agg(["func1", "func2", ...])

# Get sum, mean, and count for each region in one line
stats = df.groupby("Region")["Revenue"].agg(["sum", "mean", "count"])
print(stats)

# Result is a DataFrame with:
# - Index: the group labels (Region names)
# - Columns: each aggregation function you specified
```

## 3. Multi-Column GroupBy

Group by multiple columns for detailed analysis.

```python
# Demo: Multi-Column GroupBy
# Group by MULTIPLE columns for detailed breakdown
# Syntax: df.groupby(["col1", "col2"])["value"].aggregation()

import pandas as pd

df = pd.DataFrame({
    "Region": ["EMEA", "AMER", "EMEA", "APAC", "AMER", "EMEA"],
    "Product": ["Suite A", "Suite A", "Suite B", "Suite A", "Suite B", "Suite A"],
    "Revenue": [15000, 12000, 8500, 9500, 11000, 7500]
})

# Group by BOTH Region AND Product
# Creates groups like: (EMEA, Suite A), (EMEA, Suite B), (AMER, Suite A), etc.
detailed = df.groupby(["Region", "Product"])["Revenue"].sum()
print("Revenue by Region and Product:")
print(detailed)

# Reset index to get a regular DataFrame (easier to work with)
# The group columns become regular columns again
detailed_df = df.groupby(["Region", "Product"])["Revenue"].sum().reset_index()
print("\nAs DataFrame:")
print(detailed_df)
```

## 4. Transform - Add Group Statistics Back

`transform()` adds group calculations back to each row.

```python
# Demo: Transform - Add Group Statistics Back to Each Row
# transform() applies group calculations but keeps original row count
# Unlike agg() which returns one row per group, transform() returns same shape
# Syntax: df.groupby("column")["value"].transform("function")

import pandas as pd

df = pd.DataFrame({
    "Region": ["EMEA", "AMER", "EMEA", "APAC", "AMER"],
    "Salesperson": ["Alice", "Bob", "Charlie", "Diana", "Eve"],
    "Revenue": [15000, 12000, 8500, 9500, 11000]
})

# Add regional total to EACH row (not just one row per region)
# transform() broadcasts the group result back to each original row
df["Regional_Total"] = df.groupby("Region")["Revenue"].transform("sum")

# Add regional average
df["Regional_Avg"] = df.groupby("Region")["Revenue"].transform("mean")

# Now calculate each person's share of their regional total
df["Regional_Share"] = df["Revenue"] / df["Regional_Total"] * 100

print("Individual Performance vs Regional:")
print(df)

# Identify who performed above their regional average
df["Above_Average"] = df["Revenue"] > df["Regional_Avg"]
print("\nAbove regional average:")
print(df[df["Above_Average"]][["Salesperson", "Region", "Revenue", "Regional_Avg"]])
```
