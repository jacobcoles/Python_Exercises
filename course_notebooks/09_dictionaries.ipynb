## 1. Dictionary Basics**Dictionaries** store data as key-value pairs.Unlike lists (indexed by position), dictionaries use meaningful keys.Perfect for:- Configuration settings- Records with named fields- Lookup tables- Structured business data

````python
# Demo 1: Creating and accessing dictionaries
# Dictionaries use curly braces {} with key: value pairs
# Syntax: {key1: value1, key2: value2, ...}

product = {
    "name": "Suite A",       # String value
    "price": 199.00,         # Float value
    "region": "EMEA",        # String value
    "units_sold": 150        # Integer value
}

# Access values using square brackets with the KEY (not index!)
# Syntax: dictionary["key"]
print(f"Product: {product['name']}")
print(f"Price: €{product['price']}")

# Can use dictionary values in calculations
print(f"Revenue: €{product['price'] * product['units_sold']:,.2f}")
````

## 2. Safe Access with .get() and Key Checking

What happens if a key doesn't exist? `.get()` provides safe access with default values.

````python
# Demo 1: Safe access with .get()
# Problem: Direct access crashes if key doesn't exist

product = {
    "name": "Suite A",
    "price": 199.00,
    "region": "EMEA"
}

# This works - key exists
print(f"Name: {product['name']}")

# This would CRASH with KeyError:
# print(product['discount'])  # KeyError: 'discount'

# SOLUTION: Use .get() for safe access
# Syntax: dictionary.get("key", default_value)
# Returns default_value if key doesn't exist (instead of crashing)
discount = product.get("discount", 0)  # Returns 0 since "discount" doesn't exist
print(f"Discount: {discount}%")

# .get() without default returns None if key missing
category = product.get("category")
print(f"Category: {category}")  # None
````

````python
# Demo 2: Check if key exists with 'in' operator
# Use before accessing to avoid errors

product = {
    "name": "Suite A",
    "price": 199.00,
    "region": "EMEA"
}

# 'in' checks if a KEY exists in the dictionary
# Returns True or False
if "discount" in product:
    print(f"Discount available: {product['discount']}%")
else:
    print("No discount available")

# Common pattern: loop through fields and check each
print("\nChecking all possible fields:")
fields_to_check = ["name", "price", "discount", "category"]
for field in fields_to_check:
    if field in product:
        print(f"  {field}: {product[field]}")
    else:
        print(f"  {field}: (not set)")
````

---## 2. Iterating Through DictionariesLoop through keys, values, or both using `.items()`

````python
# Demo: Looping through dictionaries
# Use .items() to get both key and value at once

regional_revenue = {
    "EMEA": 125000,
    "AMER": 98000,
    "APAC": 67000
}

# .items() returns key-value pairs that you can unpack
# Syntax: for key, value in dictionary.items():
print("Revenue by Region:")
for region, revenue in regional_revenue.items():
    print(f"  {region}: €{revenue:,}")

# Other iteration methods:
# .keys() - just the keys
# .values() - just the values
````

## 4. Lists of Dictionaries - Tabular Data

**Very important pattern!** A list of dictionaries represents table-like data, where each dictionary is a row.

````python
# Demo: List of dictionaries - Tabular Data
# VERY IMPORTANT pattern: each dictionary is like a row in a table
# This is how real data often looks before converting to Pandas

sales = [
    {"product": "Suite A", "region": "EMEA", "units": 150, "revenue": 29850},
    {"product": "Suite B", "region": "AMER", "units": 200, "revenue": 19800},
    {"product": "Suite A", "region": "APAC", "units": 75, "revenue": 14925},
    {"product": "Suite B", "region": "EMEA", "units": 180, "revenue": 17820}
]

# Loop through each record (each dictionary in the list)
print("All Sales:")
for sale in sales:
    print(f"  {sale['product']} in {sale['region']}: {sale['units']} units = €{sale['revenue']:,}")

# Calculate totals using list comprehension
# [sale["revenue"] for sale in sales] extracts revenue from each dict
total_revenue = sum(sale["revenue"] for sale in sales)
total_units = sum(sale["units"] for sale in sales)
print(f"\nTotal: {total_units} units, €{total_revenue:,}")

# Filter: keep only records matching a condition
emea_sales = [sale for sale in sales if sale["region"] == "EMEA"]
print(f"\nEMEA sales: {len(emea_sales)} transactions")
````

---## 3. Nested DictionariesDictionaries can contain other dictionaries for hierarchical data.

````python
# Demo: Nested dictionaries
# Dictionaries can contain other dictionaries for hierarchical data
# Like a folder structure: outer key → inner dictionary → inner keys

products = {
    "SUITE-A": {                    # Outer key
        "name": "Suite A",          # Inner keys and values
        "price": 199,
        "stock": 150
    },
    "SUITE-B": {                    # Another outer key
        "name": "Suite B",
        "price": 99,
        "stock": 200
    }
}

# Access nested values: chain the keys
# Syntax: dictionary[outer_key][inner_key]
print(f"Suite A price: €{products['SUITE-A']['price']}")
print(f"Suite B stock: {products['SUITE-B']['stock']} units")
````
