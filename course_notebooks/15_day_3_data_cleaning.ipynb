## 1. Handling Missing DataReal data has gaps. Handle them properly.

```python
# Demo: Detecting and Handling Missing Values
# Missing values appear as NaN (Not a Number) or None
# Real-world data almost always has missing values!

import pandas as pd
import numpy as np  # numpy provides np.nan for missing values

# Create DataFrame with missing value
df = pd.DataFrame({
    "Product": ["A", "B", "C", "D"],
    "Price": [199, np.nan, 299, 99]  # np.nan represents missing data
})

# Detect missing values with .isna() or .isnull() (same thing)
# Returns True/False for each cell
print("Missing values (True = missing):")
print(df.isna())

# Fill missing values with fillna()
# Syntax: df["column"].fillna(value)
df["Price"] = df["Price"].fillna(0)  # Replace NaN with 0
print("\nAfter filling with 0:")
print(df)
```

### Exercise: Clean sales data

```python
# Exercise: Clean Sales Data with Multiple Missing Values
# Task: Fill different columns with appropriate values

import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Region": ["EMEA", "AMER", None, "APAC"],     # None = missing string
    "Revenue": [125000, np.nan, 67000, 98000]     # np.nan = missing number
})

# Fill missing values - use a dictionary to fill different columns differently
# Syntax: df.fillna({"column1": value1, "column2": value2})
# - Region (text): fill with "Unknown"
# - Revenue (number): fill with 0
df = df.fillna({"Region": "Unknown", "Revenue": 0})
print(df)
```

## 2. Advanced Missing Value Handling

Different strategies for different situations.

```python
# Demo: Advanced Missing Value Strategies
# Different situations call for different strategies

import pandas as pd
import numpy as np

# Sample data with multiple missing values
df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [10000, np.nan, 12000, np.nan, np.nan, 15000],
    "Region": ["EMEA", "EMEA", None, "EMEA", "EMEA", "EMEA"]
})

print("Original data:")
print(df)

# Strategy 1: Fill with a constant value
# Good for: when a default makes business sense
df_constant = df.copy()
df_constant["Sales"] = df_constant["Sales"].fillna(0)
print("\nStrategy 1 - Filled with 0:")
print(df_constant)

# Strategy 2: Fill with mean/median
# Good for: numeric data where average makes sense
df_mean = df.copy()
mean_value = df["Sales"].mean()  # Calculate mean of non-null values
df_mean["Sales"] = df_mean["Sales"].fillna(mean_value)
print(f"\nStrategy 2 - Filled with mean ({mean_value:,.0f}):")
print(df_mean)

# Strategy 3: Forward fill (ffill)
# Good for: time series where last value carries forward
df_ffill = df.copy()
df_ffill["Sales"] = df_ffill["Sales"].ffill()  # Use previous value
print("\nStrategy 3 - Forward filled:")
print(df_ffill)

# Strategy 4: Drop rows with missing values
# Good for: when you have enough data and can afford to lose some
df_dropped = df.dropna()  # Removes any row with ANY missing value
print("\nStrategy 4 - Rows with no missing values:")
print(df_dropped)
```

## 2. Removing DuplicatesClean up duplicate records.

```python
# Demo: Removing Duplicate Rows
# Duplicates can skew analysis - remove them!
# Syntax: df.drop_duplicates()

df = pd.DataFrame({
    "Product": ["A", "B", "A", "C"],  # "A" appears twice
    "Price": [199, 99, 199, 299]       # Same data for both "A" rows
})

print("Original (with duplicate):")
print(df)

# Remove duplicate rows (keeps first occurrence by default)
df_clean = df.drop_duplicates()
print("\nAfter removing duplicates:")
print(df_clean)

# Options:
# df.drop_duplicates(keep='first')  # Keep first occurrence (default)
# df.drop_duplicates(keep='last')   # Keep last occurrence
# df.drop_duplicates(subset=['Product'])  # Only check specific columns
```
