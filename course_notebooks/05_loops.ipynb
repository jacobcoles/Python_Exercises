## 1. Why Loops?

Loops let you repeat operations without writing the same code multiple times. Essential for processing lists of data!

Imagine calculating revenue for 100 products. You wouldn't want to write:
```python
revenue_1 = price_1 * units_1
revenue_2 = price_2 * units_2
revenue_3 = price_3 * units_3
# ... 97 more lines!
```

Instead, use a loop!

### Demo: The Problem Loops Solve

````python
# Demo: The Problem Loops Solve
# Without loops, you have to write repetitive code for each item

# Manual approach - NOT scalable!
product_1 = "Basic"
product_2 = "Plus"
product_3 = "Pro"

print(f"Product: {product_1}")
print(f"Product: {product_2}")
print(f"Product: {product_3}")

print("\n--- This is not scalable! ---\n")

# With a loop - clean, simple, handles ANY number of items
# Syntax: for item in collection:
#             code that uses item
products = ["Basic", "Plus", "Pro", "Enterprise", "Analytics"]

# The 'for' loop runs the indented code once for each item
# 'product' is a variable that takes the value of each item in turn
for product in products:
    print(f"Product: {product}")
````

## 2. Basic For Loops

The `for` loop iterates over items in a collection (like a list).

**Syntax:**
```python
for item in collection:
    # code to run for each item
```

### Demo: Iterating Over Lists

````python
# Demo 1: Iterate over product names
# 'for' loops through each item in a list, one at a time

products = ["Basic", "Plus", "Pro", "Enterprise", "Analytics"]

# 'product' is a loop variable - it holds the current item
# Loop runs 5 times: product = "Basic", then "Plus", then "Pro", etc.
for product in products:
    print(f"Product: {product}")
````

````python
# Demo 2: Process region codes
# The loop variable name can be anything meaningful

regions = ["AMER", "EMEA", "APAC", "NORD", "LATAM"]

# 'region' will be each string in the list, one at a time
# The loop runs once per item in the list
for region in regions:
    print(f"Processing data for region: {region}")
````

````python
# Demo 3: Calculate total revenue using a loop
# Common pattern: start with a value, update it in each iteration

revenues = [3082.59, 556.34, 2422.68, 4155.00, 2486.50]

# Step 1: Initialize a "running total" variable BEFORE the loop
total_revenue = 0

# Step 2: Loop through each revenue and add it to the total
for revenue in revenues:
    total_revenue = total_revenue + revenue  # Add current value to running total

print(f"Total Revenue: €{total_revenue:,.2f}")

# Shorthand: += operator adds to the variable
# x += y is the same as x = x + y
total = 0
for revenue in revenues:
    total += revenue  # Same as: total = total + revenue

print(f"Total (using +=): €{total:,.2f}")
````

## 3. Range() Function

`range()` generates sequences of numbers - perfect for loops!

**Syntax:**
- `range(stop)` - from 0 to stop-1
- `range(start, stop)` - from start to stop-1
- `range(start, stop, step)` - with custom increment

### Demo: Using Range()

````python
# Demo 1: Basic range()
# range(n) generates numbers from 0 to n-1
# range(5) gives: 0, 1, 2, 3, 4

print("Processing 5 records:")
for i in range(5):  # i will be 0, 1, 2, 3, 4
    # Add 1 to display human-friendly numbering (1, 2, 3, 4, 5)
    print(f"Record {i+1}")
````

````python
# Demo 2: range() with start and stop
# range(start, stop) generates numbers from start to stop-1
# The stop value is NOT included (just like slicing)

print("Years 2020-2024:")
for year in range(2020, 2025):  # 2020, 2021, 2022, 2023, 2024
    print(f"Processing data for year: {year}")
    
# Note: range(2020, 2025) includes 2020, 2021, 2022, 2023, 2024
# The end value 2025 is NOT included
````

````python
# Demo 3: range() with step (increment)
# range(start, stop, step) - step controls the increment between values

print("Quarterly analysis:")
# range(1, 5) gives: 1, 2, 3, 4 (perfect for Q1, Q2, Q3, Q4)
for quarter in range(1, 5):
    print(f"Q{quarter} analysis complete")

print("\nEvery other month:")
# range(1, 13, 2) gives: 1, 3, 5, 7, 9, 11 (step of 2)
for month in range(1, 13, 2):
    print(f"Month {month}")
````

````python
# Demo 4: Using range() with len() to access list by index
# Common pattern: range(len(list)) to get indices 0, 1, 2, ...

monthly_revenue = [15000, 18000, 22000, 19000, 21000]

cumulative = 0  # Running total

# range(len(monthly_revenue)) gives: 0, 1, 2, 3, 4
# We use i as an index to access list items
for i in range(len(monthly_revenue)):
    # Use i to get item from list: monthly_revenue[i]
    cumulative += monthly_revenue[i]
    
    # i+1 for human-friendly month numbers (1, 2, 3...)
    print(f"Month {i+1}: €{monthly_revenue[i]:,} (Cumulative: €{cumulative:,})")
````

## 4. Enumerate() - Track Positions

`enumerate()` gives you both the index and the value - perfect when you need both!

**Syntax:**
```python
for index, value in enumerate(list):
    # index is the position (0, 1, 2, ...)
    # value is the item
```

### Demo: Using Enumerate()

````python
# Demo 1: enumerate() for numbered list
# enumerate() gives you BOTH the index AND the value at once

products = ["Basic", "Plus", "Pro", "Enterprise"]

# Without enumerate, you'd need range(len(products)) and products[i]
# With enumerate: for index, value in enumerate(list)
for index, product in enumerate(products):
    # index = 0, 1, 2, 3...
    # product = "Basic", "Plus", "Pro", ...
    print(f"{index+1}. {product}")  # Add 1 for 1-based numbering
````

````python
# Demo 2: enumerate() with start parameter
# You can tell enumerate to start counting from a different number

products = ["Basic", "Plus", "Pro", "Enterprise"]

# start=1 makes the first index 1 instead of 0
# Perfect when you want human-friendly numbering
for number, product in enumerate(products, start=1):
    # number starts at 1, not 0
    print(f"{number}. {product}")
````

````python
# Demo 3: Using enumerate() to modify list items
# IMPORTANT: You need the index to change items in a list

prices = [49.00, 99.00, 199.00, 499.00]
print(f"Original prices: {prices}")

# Apply 10% increase to all prices
for i, price in enumerate(prices):
    # i = index, price = current value
    # Multiply by 1.10 for 10% increase
    # Assign back to prices[i] to modify the list
    prices[i] = price * 1.10

# Show updated list
print(f"Updated prices: {[f'{p:.2f}' for p in prices]}")
````

## 5. Building Lists with Loops

A common pattern: start with an empty list and add items in a loop.

### Demo: Accumulating Results

````python
# Demo 1: Build a list of calculated margins
# Pattern: Start empty, append results in loop

revenues = [3082.59, 556.34, 2422.68, 4155.00]
costs = [2500.00, 400.00, 1800.00, 3200.00]

# Step 1: Start with an empty list
margins = []

# Step 2: Loop and calculate each margin
for i in range(len(revenues)):
    # Calculate margin for this item
    margin = revenues[i] - costs[i]
    
    # Append to the results list
    margins.append(margin)

print(f"Revenues: {revenues}")
print(f"Costs: {costs}")
print(f"Margins: {margins}")
````

````python
# Demo 2: Filter items with a condition
# Pattern: Check condition, only append if True

all_amounts = [3082.59, 556.34, 2422.68, 4155.00, 2486.50, 15000.00]

# Start with empty list for filtered results
large_transactions = []

for amount in all_amounts:
    # Only add items that meet the condition
    if amount > 3000:
        large_transactions.append(amount)

print(f"All amounts: {all_amounts}")
print(f"Large transactions (>€3000): {large_transactions}")
````

````python
# Demo 3: Clean and transform data
# Pattern: Transform each item before adding

raw_product_codes = ["  p-bsc  ", "P-ENT", "p-pro", "P-ANL  "]

clean_codes = []

for code in raw_product_codes:
    # Transform the code:
    # strip() removes leading/trailing whitespace
    # upper() converts to uppercase
    clean_code = code.strip().upper()
    
    # Add the cleaned version to results
    clean_codes.append(clean_code)

print(f"Raw codes: {raw_product_codes}")
print(f"Clean codes: {clean_codes}")
````

## 6. While Loops - Repeat Until Condition Changes

`while` loops repeat as long as a condition is True.

**Syntax:**
```python
while condition:
    # code to run
    # (must eventually make condition False!)
```

### Demo: While Loops

````python
# Demo 1: Simple countdown with while loop
# while loops repeat AS LONG AS the condition is True
# IMPORTANT: You must change the condition variable inside the loop!

countdown = 5

# Loop continues while countdown > 0
while countdown > 0:
    print(f"Countdown: {countdown}")
    countdown -= 1  # Decrease by 1 each time (makes condition eventually False)

print("Liftoff!")

# WARNING: Without 'countdown -= 1', this would run forever (infinite loop)!
````

````python
# Demo 2: Business example - accumulate until target reached
# while is useful when you don't know how many iterations you need

daily_sales = [1200, 1450, 980, 1680, 1320, 1890, 2100, 1560]
target = 5000
total = 0
day = 0

# Two conditions combined with 'and':
# 1. total < target - haven't reached the goal yet
# 2. day < len(daily_sales) - still have data to process
while total < target and day < len(daily_sales):
    total += daily_sales[day]  # Add this day's sales
    day += 1                   # Move to next day
    
    # day-1 because we already incremented day
    print(f"Day {day}: Sales €{daily_sales[day-1]:,}, Running total: €{total:,}")

# Report whether target was reached
if total >= target:
    print(f"\nTarget reached on Day {day}!")
else:
    print(f"\nTarget not reached. Total after {day} days: €{total:,}")
````

## 7. Break and Continue - Loop Control

Sometimes you need more control over loops:
- `break` - Exit the loop immediately
- `continue` - Skip to the next iteration

### Demo: Break and Continue

````python
# Demo 1: break - exit the loop immediately when found
# Use break when you're searching and only need the first match

transactions = [1500, 800, 2200, 5500, 1200, 3400]

print("Searching for first transaction over €5000...")

for amount in transactions:
    print(f"Checking €{amount:,}")
    
    if amount > 5000:
        print(f"Found it! €{amount:,}")
        break  # Exit the loop immediately - don't check remaining items

print("Search complete")
# Notice: We stopped at 5500 and didn't check 1200 or 3400
````

````python
# Demo 2: continue - skip current item and move to next
# Use continue when you want to ignore certain items

revenues = [3500, -100, 2800, 0, 4200, -50, 1900]

print("Processing only valid revenues (positive numbers):")
total = 0

for revenue in revenues:
    # Skip invalid values (zero or negative)
    if revenue <= 0:
        print(f"Skipping invalid: €{revenue}")
        continue  # Jump to next iteration, skip code below
    
    # This code only runs for valid (positive) revenues
    total += revenue
    print(f"Added €{revenue:,} - Running total: €{total:,}")

print(f"\nTotal valid revenue: €{total:,}")
````

## 8. List Comprehensions (Optional/Advanced)

List comprehensions are a concise way to create lists - very "Pythonic"!

*Note: This is an advanced technique. It's fine to use regular loops instead!*

**Syntax:**
```python
[expression for item in list]  # Transform
[expression for item in list if condition]  # Filter and transform
```

### Demo: List Comprehensions

````python
# Demo 1: List comprehension vs traditional loop
# Comprehensions create lists in a single, concise line

prices = [49, 99, 199, 499]

# Traditional way with loop
increased_prices_loop = []
for price in prices:
    increased_prices_loop.append(price * 1.10)

# List comprehension - same result in one line!
# Syntax: [expression for item in list]
# Reads as: "price * 1.10 for each price in prices"
increased_prices_comp = [price * 1.10 for price in prices]

print(f"Original: {prices}")
print(f"Loop result: {increased_prices_loop}")
print(f"Comprehension result: {increased_prices_comp}")
````

````python
# Demo 2: Filter with list comprehension
# Add 'if condition' at the end to filter items

amounts = [3082.59, 556.34, 2422.68, 4155.00, 2486.50]

# Syntax: [expression for item in list if condition]
# Only includes items where condition is True
large = [amt for amt in amounts if amt > 3000]

print(f"All amounts: {amounts}")
print(f"Large amounts: {large}")
````

````python
# Demo 3: String transformations with comprehension
# Apply a method to each item in the list

raw_codes = ["  p-bsc  ", "P-ENT", "p-pro"]

# Transform each code: remove whitespace and convert to uppercase
# code.strip() removes spaces, .upper() makes uppercase
clean = [code.strip().upper() for code in raw_codes]

print(f"Raw: {raw_codes}")
print(f"Clean: {clean}")
````

### When to Use Comprehensions

**Use comprehensions for:**
- Simple transformations
- Filtering
- Creating new lists from existing ones

**Use regular loops for:**
- Complex logic with multiple steps
- When you need to debug
- When readability would suffer
