## 1. Boolean FilteringFilter rows based on conditions.

```python
# Demo: Basic Boolean Filtering
# Filter rows based on a condition
# Syntax: df[df["column"] comparison_operator value]

import pandas as pd

# Create sample data
df = pd.DataFrame({
    "Product": ["Suite A", "Suite B", "Suite C"],
    "Price": [199, 99, 299],
    "Stock": [150, 45, 200]
})

# Filter for products where Price > 150
# Step 1: df["Price"] > 150 creates True/False for each row
# Step 2: df[...] keeps only rows where condition is True
expensive = df[df["Price"] > 150]
print(expensive)
```

## 2. Combining ConditionsUse & (and) and | (or) to combine filters.

```python
# Demo: Combining Conditions with AND (&)
# Use & when BOTH conditions must be True
# CRITICAL: Wrap EACH condition in parentheses!
# Syntax: df[(condition1) & (condition2)]

# Find products with high price AND low stock (critical items!)
critical = df[(df["Price"] > 100) & (df["Stock"] < 50)]
print(critical)
```

## 3. Filtering with isin() - Multiple Values

Use `isin()` when checking for multiple possible values.

```python
# Demo: Filtering with isin() - Multiple Values at Once
# isin() checks if values are in a list - cleaner than multiple OR conditions
# Syntax: df[df["column"].isin([value1, value2, value3])]

import pandas as pd

df = pd.DataFrame({
    "Product": ["Suite A", "Suite B", "Suite C", "Suite D", "Suite E"],
    "Region": ["EMEA", "AMER", "APAC", "EMEA", "NORD"],
    "Price": [199, 99, 299, 149, 249]
})

# Define list of priority regions
priority_regions = ["EMEA", "AMER"]

# Filter using isin() - much cleaner than: (df["Region"] == "EMEA") | (df["Region"] == "AMER")
priority_products = df[df["Region"].isin(priority_regions)]

print("Priority regions (EMEA, AMER):")
print(priority_products)

# Negation: Use ~ to get rows NOT in the list
# Syntax: df[~df["column"].isin(list)]
other_regions = df[~df["Region"].isin(priority_regions)]
print("\nOther regions:")
print(other_regions)
```

## 4. String Filtering with str.contains()

Filter text columns using string patterns.

```python
# Demo: String Filtering with str.contains()
# Use .str accessor to apply string methods to entire columns
# str.contains() checks if substring exists in each value
# Syntax: df[df["column"].str.contains("pattern")]

import pandas as pd

df = pd.DataFrame({
    "Product": ["Enterprise Suite", "Basic Plan", "Pro Suite", "Enterprise Plan", "Starter"],
    "Code": ["ENT-001", "BAS-002", "PRO-003", "ENT-004", "STR-005"],
    "Price": [499, 49, 199, 399, 29]
})

# Filter products containing "Suite" in the name
suites = df[df["Product"].str.contains("Suite")]
print("Products with 'Suite' in name:")
print(suites)

# Case-insensitive search using case=False parameter
enterprise = df[df["Product"].str.contains("enterprise", case=False)]
print("\nEnterprise products (case-insensitive):")
print(enterprise)

# Filter by code prefix using str.startswith()
# Syntax: df[df["column"].str.startswith("prefix")]
ent_codes = df[df["Code"].str.startswith("ENT")]
print("\nProducts with 'ENT' code prefix:")
print(ent_codes)
```
