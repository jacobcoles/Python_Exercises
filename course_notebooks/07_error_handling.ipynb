## 2. Basic Try/Except

**Syntax:**
```python
try:
    # Code that might raise an error
    risky_operation()
except ErrorType:
    # Code to run if error occurs
    handle_error()
```

### Demo: Safe Data Conversion

````python
# Demo 1: Basic try/except for safe data conversion
# try/except prevents crashes by catching errors

values = ["100", "200", "invalid", "300"]

total = 0
errors = 0

for value in values:
    # 'try' block contains code that might raise an error
    try:
        # This might fail if value can't be converted to int
        number = int(value)
        total += number
    
    # 'except' block runs when a specific error type occurs
    # ValueError occurs when conversion fails (e.g., int("invalid"))
    except ValueError:
        # Handle the error gracefully - don't crash
        print(f"Warning: Cannot convert '{value}' to number")
        errors += 1

# Program continues even though some values failed
print(f"\nTotal: {total}")
print(f"Errors encountered: {errors}")
print(f"Successfully processed: {len(values) - errors} values")
````

````python
# Demo 2: Safe division - handling ZeroDivisionError
# Division by zero is a common error that crashes programs

def calculate_margin(revenue, cost):
    """
    Calculate profit margin with error handling.
    Returns error message if revenue is zero.
    """
    try:
        # This will fail if revenue is 0
        profit = revenue - cost
        margin = (profit / revenue) * 100
        return margin
    
    # ZeroDivisionError occurs when dividing by zero
    except ZeroDivisionError:
        # Return a meaningful error message instead of crashing
        return "Error: Revenue cannot be zero"

# Test with normal values - works fine
print(f"Normal: {calculate_margin(1000, 600):.1f}%")

# Test with zero revenue - handled gracefully
print(f"Zero revenue: {calculate_margin(0, 600)}")
````

## 3. Handling Multiple Error Types

Different operations can raise different errors. Handle them separately for better error messages.

### Demo: Multiple Except Blocks

````python
# Demo 1: Handle different error types with multiple except blocks
# Different operations can raise different types of errors

def safe_calculate(revenue_str, cost_str):
    """
    Calculate margin with multiple error handlers.
    Each error type gets its own message.
    """
    try:
        # Multiple things can go wrong here:
        # 1. float() might fail if strings aren't numbers
        revenue = float(revenue_str)
        cost = float(cost_str)
        
        # 2. Division might fail if revenue is 0
        margin = ((revenue - cost) / revenue) * 100
        return f"Margin: {margin:.1f}%"
    
    # ValueError: conversion failures (e.g., float("invalid"))
    except ValueError:
        return "Error: Invalid number format"
    
    # ZeroDivisionError: division by zero
    except ZeroDivisionError:
        return "Error: Revenue cannot be zero"
    
    # TypeError: wrong types (e.g., float(None))
    except TypeError:
        return "Error: Values must be numbers"

# Test different scenarios - each triggers different error handler
print(safe_calculate("1000", "600"))      # Normal - no error
print(safe_calculate("invalid", "600"))   # ValueError - can't convert
print(safe_calculate("0", "600"))         # ZeroDivisionError - divide by 0
print(safe_calculate(None, "600"))        # TypeError - None isn't a string
````

````python
# Demo 2: Catch multiple error types in one except block
# Use a tuple (Error1, Error2) to catch several types together

def process_transaction(amount_str):
    """
    Process transaction with grouped error handling.
    """
    try:
        # Convert string to float
        amount = float(amount_str)
        
        # 'raise' manually triggers an error
        # We use it for business validation
        if amount <= 0:
            raise ValueError("Amount must be positive")
        
        return f"Processed: €{amount:,.2f}"
    
    # Catch BOTH ValueError and TypeError with one except
    # Use 'as e' to capture the error message
    except (ValueError, TypeError) as e:
        return f"Error: {e}"

# Test cases
print(process_transaction("1500"))      # Valid - works
print(process_transaction("invalid"))   # ValueError from float()
print(process_transaction("-100"))      # ValueError from our raise statement
````

### Exercise 2: Safe Data Parser

Create a function `parse_sales_record()` that:
- Takes a string: "ProductID,Revenue,Units"
- Returns a dictionary: {"product_id": str, "revenue": float, "units": int}
- Handles ValueError if conversion fails
- Handles IndexError if not enough fields
- Returns None on any error

Test with:
- "PROD-001,1500.50,150" (valid)
- "PROD-002,invalid,200" (ValueError)
- "PROD-003,2000" (IndexError - missing field)

````python
# Exercise 2: Safe Data Parser
# This example is already complete - study how it handles multiple error types

def parse_sales_record(record_str):
    """
    Parse sales record string into dictionary.
    Returns None if parsing fails.
    
    Handles:
    - ValueError: if number conversion fails
    - IndexError: if not enough fields in the string
    """
    try:
        # Split the string by comma into a list
        # "PROD-001,1500.50,150" → ["PROD-001", "1500.50", "150"]
        parts = record_str.split(",")
        
        # Access each part by index - can raise IndexError
        product_id = parts[0]
        
        # Convert to numbers - can raise ValueError
        revenue = float(parts[1])
        units = int(parts[2])
        
        # If we got here, all conversions succeeded
        # Return a dictionary with the parsed values
        return {
            "product_id": product_id,
            "revenue": revenue,
            "units": units
        }
    
    # Catch both error types - 'as e' captures the error message
    except (ValueError, IndexError) as e:
        print(f"Parse error: {e}")
        return None

# Test with various inputs
test_records = [
    "PROD-001,1500.50,150",   # Valid - all fields present
    "PROD-002,invalid,200",   # ValueError - can't convert "invalid"
    "PROD-003,2000"           # IndexError - missing third field
]

for record in test_records:
    print(f"\nParsing: {record}")
    result = parse_sales_record(record)
    if result:
        print(f"Success: {result}")
    else:
        print("Failed to parse")
````

## 4. Try/Except/Else/Finally

Complete error handling with all clauses:
- **try**: Code that might raise an error
- **except**: Handle the error
- **else**: Runs only if no error occurred
- **finally**: Always runs (cleanup code)

### Demo: Complete Error Handling

````python
# Demo: Complete error handling with try/except/else/finally
# These four blocks give you full control over error handling

def process_transaction(amount_str):
    """
    Process transaction with complete error handling.
    
    Structure:
    - try: Code that might fail
    - except: Handle errors
    - else: Runs ONLY if no error occurred
    - finally: ALWAYS runs (for cleanup)
    """
    try:
        # Try to convert and validate
        amount = float(amount_str)
        
        # Manually raise error for business validation
        if amount <= 0:
            raise ValueError("Amount must be positive")
    
    except ValueError as e:
        # Handle conversion or validation errors
        print(f"Error: {e}")
        return None
    
    else:
        # Only runs if try block completed WITHOUT errors
        # Use this for success-only operations
        print(f"Transaction processed: €{amount:,.2f}")
        return amount
    
    finally:
        # Always runs, whether there was an error or not
        # Use for cleanup, logging, closing files, etc.
        print("Transaction attempt logged\n")

# Test 1: Valid input - try succeeds, else runs, finally runs
print("Test 1: Valid")
process_transaction("1500")

# Test 2: Invalid format - try fails, except runs, finally runs
print("Test 2: Invalid format")
process_transaction("invalid")

# Test 3: Negative - try raises error, except runs, finally runs
print("Test 3: Negative amount")
process_transaction("-100")
````

### Exercise 3: File Processing Simulator

Create a function `process_data_file()` that:
- Takes a list of data strings
- Tries to convert each to float
- Keeps track of successful and failed conversions
- Uses else to add to success list
- Uses finally to print summary

Test with: ["100.5", "200.3", "invalid", "300.7", "bad"]

````python
# Exercise 3: File Processing Simulator
# This example demonstrates processing a list with error tracking

def process_data_file(data_list):
    """
    Process list of data with complete error handling.
    Tracks which items succeeded and which failed.
    """
    successful = []  # Store converted values
    failed = []      # Store items that couldn't be converted
    
    for item in data_list:
        try:
            # Try to convert string to float
            value = float(item)
        
        except ValueError:
            # Conversion failed - add to failed list
            failed.append(item)
        
        else:
            # Conversion succeeded - add to successful list
            # else only runs when try succeeds
            successful.append(value)
    
    # Print summary (could also use finally for cleanup)
    print(f"Processed {len(data_list)} items")
    print(f"Successful: {len(successful)}")
    print(f"Failed: {len(failed)}")
    
    if successful:
        print(f"Total value: {sum(successful):.2f}")
    
    return successful, failed

# Test with mixed data - some valid, some invalid
test_data = ["100.5", "200.3", "invalid", "300.7", "bad"]
good, bad = process_data_file(test_data)

print(f"\nSuccessful values: {good}")
print(f"Failed values: {bad}")
````

## 5. Common Python Errors

Understanding error types helps you handle them appropriately.

### Demo: Common Error Types

````python
# Demo: Common Python error types and how to handle them
# Each error type indicates a specific problem

# 1. ValueError - Value is wrong type or format for conversion
try:
    number = int("not a number")  # Can't convert this to integer
except ValueError:
    print("ValueError: Cannot convert 'not a number' to integer")

# 2. ZeroDivisionError - Attempted to divide by zero
try:
    result = 10 / 0  # Division by zero is undefined
except ZeroDivisionError:
    print("ZeroDivisionError: Cannot divide by zero")

# 3. KeyError - Dictionary key doesn't exist
try:
    data = {"name": "Product A"}
    price = data["price"]  # "price" key doesn't exist
except KeyError:
    print("KeyError: Key 'price' not found in dictionary")

# 4. IndexError - List index is out of range
try:
    items = [1, 2, 3]  # Valid indices: 0, 1, 2
    value = items[10]  # Index 10 doesn't exist
except IndexError:
    print("IndexError: List index 10 out of range (list has 3 items)")

# 5. TypeError - Wrong type for an operation
try:
    result = "text" + 5  # Can't add string and integer
except TypeError:
    print("TypeError: Cannot add string and integer (use str(5) first)")
````

## Challenge Exercise: Robust Data Cleaner

Build a complete data cleaning function that handles all common errors.

**Task:** Create `clean_sales_data()` that:
1. Takes a list of dictionaries: [{"product": str, "revenue": str, "units": str}]
2. Converts revenue to float and units to int
3. Validates that revenue and units are positive
4. Handles missing keys gracefully
5. Returns two lists: valid_records and errors

**Test Data:**
```python
[
    {"product": "Suite A", "revenue": "1500.50", "units": "150"},
    {"product": "Suite B", "revenue": "invalid", "units": "200"},
    {"product": "Suite C", "revenue": "2000"},  # Missing units
    {"product": "Suite D", "revenue": "-500", "units": "100"},  # Negative
    {"product": "Suite E", "revenue": "3000.00", "units": "250"}
]
```

````python
# Challenge Exercise: Robust Data Cleaner
# Complete example showing comprehensive error handling for data cleaning

def clean_sales_data(raw_data):
    """
    Clean and validate sales data with comprehensive error handling.
    
    Parameters:
    raw_data (list): List of dictionaries with product, revenue, units
    
    Returns:
    tuple: (valid_records, errors) - lists of successes and failures
    """
    valid_records = []
    errors = []
    
    # enumerate with start=1 for human-friendly record numbers
    for i, record in enumerate(raw_data, 1):
        try:
            # Step 1: Extract fields - can raise KeyError if missing
            product = record["product"]
            
            # Step 2: Convert types - can raise ValueError if invalid
            revenue = float(record["revenue"])
            units = int(record["units"])
            
            # Step 3: Business validation - manually raise ValueError
            if revenue <= 0:
                raise ValueError("Revenue must be positive")
            if units <= 0:
                raise ValueError("Units must be positive")
            
            # Step 4: If we got here, record is valid
            valid_records.append({
                "product": product,
                "revenue": revenue,
                "units": units
            })
        
        # Handle missing dictionary keys
        except KeyError as e:
            errors.append(f"Record {i}: Missing field {e}")
        
        # Handle conversion errors and validation errors
        except ValueError as e:
            errors.append(f"Record {i}: {e}")
        
        # Catch-all for unexpected errors
        except Exception as e:
            errors.append(f"Record {i}: Unexpected error - {e}")
    
    return valid_records, errors

# Test data with various problems
raw_sales = [
    {"product": "Suite A", "revenue": "1500.50", "units": "150"},  # Valid
    {"product": "Suite B", "revenue": "invalid", "units": "200"},  # Bad revenue
    {"product": "Suite C", "revenue": "2000"},                     # Missing units
    {"product": "Suite D", "revenue": "-500", "units": "100"},     # Negative revenue
    {"product": "Suite E", "revenue": "3000.00", "units": "250"}   # Valid
]

# Clean the data
valid, errors = clean_sales_data(raw_sales)

# Print comprehensive report
print("=" * 50)
print("DATA CLEANING REPORT")
print("=" * 50)
print(f"\nTotal records processed: {len(raw_sales)}")
print(f"Valid records: {len(valid)}")
print(f"Invalid records: {len(errors)}")

if valid:
    print("\nValid Records:")
    for record in valid:
        print(f"  {record['product']}: €{record['revenue']:,.2f} ({record['units']} units)")

if errors:
    print("\nErrors:")
    for error in errors:
        print(f"  ✗ {error}")

print("=" * 50)
````
