## 2. Writing Clean, Readable Code

### Code is read more than it's written!

### ❌ Bad Code:

````python
# BAD CODE EXAMPLE: Unclear and Hard to Maintain
# Problems: 
# - Single-letter variable names (d, x, y) - what do they mean?
# - No comments explaining purpose
# - Abbreviations in column names ('r', 'p') - confusing

import pandas as pd
d=pd.read_csv('data.csv')       # What is 'd'? What data?
x=d[d['r']>1000]                # What is 'r'? Why 1000?
y=x.groupby('p')['r'].sum()     # What is 'p'? What are we calculating?
print(y)

# This code works, but:
# - You won't understand it in 2 weeks
# - Others can't maintain it
# - Bugs are hard to find
````

### ✅ Good Code:

````python
# GOOD CODE EXAMPLE: Clear and Maintainable
# Key improvements:
# - Descriptive variable names (sales_df, high_value_sales, product_revenue)
# - Comments explaining the "why" not just the "what"
# - Proper spacing for readability

import pandas as pd

# Load sales data from CSV file
sales_df = pd.read_csv('sales_data.csv')

# Filter high-value transactions (above €1000 threshold for premium analysis)
high_value_sales = sales_df[sales_df['revenue'] > 1000]

# Calculate total revenue by product for sales report
product_revenue = high_value_sales.groupby('product')['revenue'].sum()

print("Revenue by product:")
print(product_revenue)

# This code is self-documenting:
# - Clear names tell you what each variable contains
# - Comments explain business logic (why €1000?)
# - Easy to debug and modify later
````

## 3. Using Functions Effectively

### Don't Repeat Yourself (DRY Principle)

### ❌ Repetitive Code:

````python
# BAD EXAMPLE: Copy-Pasted Repetitive Code
# Problems:
# - Same logic repeated 3 times (DRY violation)
# - If you need to change the calculation, you must change it in 3 places
# - Easy to introduce bugs when copying

import pandas as pd

# Analyze EMEA region
emea = df[df['region'] == 'EMEA']
emea_total = emea['revenue'].sum()
emea_avg = emea['revenue'].mean()
print(f"EMEA: Total={emea_total}, Avg={emea_avg}")

# Analyze AMER region (copy-pasted, changed one word)
amer = df[df['region'] == 'AMER']
amer_total = amer['revenue'].sum()
amer_avg = amer['revenue'].mean()
print(f"AMER: Total={amer_total}, Avg={amer_avg}")

# Analyze APAC region (copy-pasted again)
apac = df[df['region'] == 'APAC']
apac_total = apac['revenue'].sum()
apac_avg = apac['revenue'].mean()
print(f"APAC: Total={apac_total}, Avg={apac_avg}")

# What if you need to add median? You'd have to change ALL THREE blocks!
````

### ✅ Use Functions:

````python
# GOOD EXAMPLE: Use Functions to Avoid Repetition (DRY Principle)
# DRY = Don't Repeat Yourself
# Write once, use many times

def analyze_region(df, region_name):
    """
    Calculate total and average revenue for a region.
    
    Parameters:
    df (DataFrame): Sales data with 'region' and 'revenue' columns
    region_name (str): Region to analyze (e.g., 'EMEA', 'AMER')
    
    Returns:
    dict: Dictionary with 'total' and 'average' revenue
    """
    # Filter to specific region
    region_data = df[df['region'] == region_name]
    
    # Calculate and return metrics
    return {
        'total': region_data['revenue'].sum(),
        'average': region_data['revenue'].mean()
    }

# Now use the function for all regions - ONE loop, not copy-paste
for region in ['EMEA', 'AMER', 'APAC']:
    stats = analyze_region(df, region)
    print(f"{region}: Total={stats['total']:,.2f}, Avg={stats['average']:,.2f}")

# Benefits:
# - Change calculation in ONE place, applies everywhere
# - Easy to add new regions (just add to list)
# - Function is reusable in other analyses
````

### Docstrings for Functions:

````python
# BEST PRACTICE: Docstrings for Functions
# A docstring documents what a function does, its parameters, and return value
# Triple quotes allow multi-line documentation

def calculate_growth_rate(current, previous):
    """
    Calculate percentage growth between two values.
    
    Parameters:
    current (float): Current period value (e.g., this year's revenue)
    previous (float): Previous period value (e.g., last year's revenue)
    
    Returns:
    float: Growth rate as a percentage (e.g., 10.0 means 10% growth)
    
    Example:
    >>> calculate_growth_rate(110, 100)
    10.0
    >>> calculate_growth_rate(90, 100)
    -10.0
    """
    # Handle edge case: avoid division by zero
    if previous == 0:
        return 0
    
    # Growth formula: ((new - old) / old) * 100
    return ((current - previous) / previous) * 100

# Docstrings are shown when you use help(function_name)
# They make your code professional and maintainable
````

## 5. Git Version Control (4 Essential Commands)

### Why Use Git?
- Track changes over time
- Collaborate with others

## 6. Common Pitfalls to Avoid

### 1. Modifying Original Data
```python
# ❌ Bad
df = pd.read_csv('sales.csv')
df = df[df['revenue'] > 1000]  # Original data lost!

# ✅ Good
df_raw = pd.read_csv('sales.csv')
df_filtered = df_raw[df_raw['revenue'] > 1000]  # Keep original!
```

### 2. Hardcoding Values
```python
# ❌ Bad
revenue_after_tax = revenue * 0.80  # What's 0.80?

# ✅ Good
TAX_RATE = 0.20
revenue_after_tax = revenue * (1 - TAX_RATE)
```

### 3. Not Handling Errors
```python
# ❌ Bad
margin = profit / revenue  # Crashes if revenue is 0!

# ✅ Good
if revenue == 0:
    margin = 0
else:
    margin = profit / revenue
```

### 4. Ignoring Data Types
```python
# ❌ Bad
df['revenue'].sum()  # Might be strings!

# ✅ Good
df['revenue'] = pd.to_numeric(df['revenue'], errors='coerce')
df['revenue'].sum()
```
