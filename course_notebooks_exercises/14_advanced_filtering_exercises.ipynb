## Setup

```python
import pandas as pd

df = pd.DataFrame({
    "Product": ["Suite A", "Suite B", "Suite C", "Suite D"],
    "Price": [199, 99, 299, 149],
    "Stock": [150, 45, 200, 15]
})

print(df)
```

## 1. Boolean Filtering

```python
# Demo: Filter where Price > 150
expensive = df[df["Price"] > 150]
print("Price > 150:")
print(expensive)
```

## 2. Combining Conditions

```python
# Demo: Combining with AND (&)
# CRITICAL: Wrap EACH condition in parentheses!

critical = df[(df["Price"] > 100) & (df["Stock"] < 50)]
print("Price > 100 AND Stock < 50:")
print(critical)
```

## 3. Filtering with isin()

```python
# Demo: isin() for multiple values
df2 = pd.DataFrame({
    "Product": ["Suite A", "Suite B", "Suite C", "Suite D", "Suite E"],
    "Region": ["EMEA", "AMER", "APAC", "EMEA", "NORD"],
    "Price": [199, 99, 299, 149, 249]
})

# Filter for EMEA or AMER
priority_regions = ["EMEA", "AMER"]
priority = df2[df2["Region"].isin(priority_regions)]
print("Priority regions:")
print(priority)
```

## 4. String Filtering

```python
# Demo: str.contains()
products = pd.DataFrame({
    "Product": ["Enterprise Suite", "Basic Plan", "Pro Suite", "Enterprise Plan"],
    "Code": ["ENT-001", "BAS-002", "PRO-003", "ENT-004"]
})

# Products containing "Suite"
suites = products[products["Product"].str.contains("Suite")]
print("Suite products:")
print(suites)
```
