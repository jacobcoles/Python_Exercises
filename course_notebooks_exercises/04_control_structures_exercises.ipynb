## 2. Elif for Multiple Conditions

```python
# Demo: Multiple conditions with elif
amount = 3500

if amount < 1000:
    category = "Small"
elif amount < 5000:
    category = "Medium"
elif amount < 10000:
    category = "Large"
else:
    category = "Very Large"

print(f"Amount: €{amount:,}")
print(f"Category: {category}")
```

## 4. Logical Operators: AND, OR, NOT

- `and` - Both must be True
- `or` - At least one must be True
- `not` - Reverses True/False

```python
# Demo: AND operator
revenue = 10000
customer_type = "Premium"

# Both conditions must be True
if revenue > 5000 and customer_type == "Premium":
    print("15% premium discount applied")
else:
    print("Standard 5% discount")
```

```python
# Demo: OR operator
region = "EMEA"

# Either condition can be True
if region == "EMEA" or region == "NORD":
    print("European region - apply GDPR rules")
else:
    print("Non-European region")
```

## 5. Nested Conditions

```python
# Demo: Nested if statements
customer_tier = "Premium"
order_value = 3000

if customer_tier == "Premium":
    print("Premium Customer")
    if order_value > 5000:
        discount = 0.20
        print("High-value order: 20% discount")
    else:
        discount = 0.15
        print("Standard order: 15% discount")
else:
    discount = 0.05
    print("Standard customer: 5% discount")
```
