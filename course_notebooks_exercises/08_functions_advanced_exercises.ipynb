## 1. Default Parameters

```python
# Demo: Default parameters
def calculate_tax(amount, rate=0.20):
    """Rate defaults to 20% if not specified."""
    return amount * rate

# Use default rate
tax1 = calculate_tax(1000)
print(f"Tax at default rate: €{tax1:.2f}")

# Override with different rate
tax2 = calculate_tax(1000, 0.25)
print(f"Tax at 25%: €{tax2:.2f}")
```

## 2. Keyword Arguments

```python
# Demo: Keyword arguments for clarity
def create_order(product, quantity, price, discount=0, priority=False):
    subtotal = quantity * price
    total = subtotal * (1 - discount)
    status = "PRIORITY" if priority else "Standard"
    return f"{product} x{quantity}: €{total:.2f} [{status}]"

# Keyword arguments - much clearer!
order = create_order(
    product="Suite A",
    quantity=5,
    price=199,
    discount=0.10,
    priority=True
)
print(order)
```

## 3. Functions with Lists

```python
# Demo: Transform pattern - apply to all
def apply_discount_to_all(prices, discount_rate):
    discounted = []
    for price in prices:
        discounted.append(price * (1 - discount_rate))
    return discounted

original = [199, 99, 299, 499]
sale = apply_discount_to_all(original, 0.15)
print(f"Original: {original}")
print(f"Sale prices: {sale}")
```

```python
# Demo: Filter pattern - keep matching
def filter_large(amounts, threshold):
    large = []
    for amount in amounts:
        if amount > threshold:
            large.append(amount)
    return large

all_amounts = [250, 1500, 75, 3200, 450]
large_only = filter_large(all_amounts, 1000)
print(f"Large (>€1000): {large_only}")
```

## 4. In-Place vs Returning New Lists

```python
# Demo: The difference

# Modifies original (in-place)
def add_tax_inplace(prices, rate=0.20):
    for i in range(len(prices)):
        prices[i] = prices[i] * (1 + rate)

# Returns new list (original unchanged)
def add_tax_new(prices, rate=0.20):
    return [price * (1 + rate) for price in prices]

# Test in-place
prices1 = [100, 200, 300]
print(f"Before: {prices1}")
add_tax_inplace(prices1)
print(f"After (changed!): {prices1}")

# Test new list
prices2 = [100, 200, 300]
with_tax = add_tax_new(prices2)
print(f"\nOriginal (unchanged): {prices2}")
print(f"New list: {with_tax}")
```
