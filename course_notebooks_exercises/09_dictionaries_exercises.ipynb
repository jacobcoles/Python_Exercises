## 2. Safe Access with .get()

```python
# Demo: Safe access with .get()
product = {
    "name": "Suite A",
    "price": 199.00
}

# Direct access crashes if key missing
# print(product['discount'])  # KeyError!

# .get() returns default value if key missing
discount = product.get("discount", 0)
print(f"Discount: {discount}%")

# Check if key exists
if "discount" in product:
    print(f"Has discount: {product['discount']}")
else:
    print("No discount available")
```

## 3. Iterating Through Dictionaries

```python
# Demo: Looping with .items()
regional_revenue = {
    "EMEA": 125000,
    "AMER": 98000,
    "APAC": 67000
}

print("Revenue by Region:")
for region, revenue in regional_revenue.items():
    print(f"  {region}: €{revenue:,}")
```

## 4. Lists of Dictionaries

```python
# Demo: List of dictionaries - tabular data
sales = [
    {"product": "Suite A", "region": "EMEA", "units": 150, "revenue": 29850},
    {"product": "Suite B", "region": "AMER", "units": 200, "revenue": 19800},
    {"product": "Suite A", "region": "APAC", "units": 75, "revenue": 14925},
]

# Loop through each record
for sale in sales:
    print(f"{sale['product']} in {sale['region']}: €{sale['revenue']:,}")

# Calculate total
total = sum(sale["revenue"] for sale in sales)
print(f"\nTotal: €{total:,}")
```

## 5. Nested Dictionaries

```python
# Demo: Nested dictionaries
products = {
    "SUITE-A": {
        "name": "Suite A",
        "price": 199,
        "stock": 150
    },
    "SUITE-B": {
        "name": "Suite B",
        "price": 99,
        "stock": 200
    }
}

# Access nested values: chain the keys
print(f"Suite A price: €{products['SUITE-A']['price']}")
print(f"Suite B stock: {products['SUITE-B']['stock']} units")
```
