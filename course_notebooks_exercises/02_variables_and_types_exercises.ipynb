## 3. Strings and F-strings

Strings are text data. F-strings let you embed variables in text.

```python
# Demo: F-strings with formatting
revenue = 4987.88
large_number = 1250000
margin = 0.2534

# ,.2f = commas + 2 decimal places
print(f"Revenue: €{revenue:,.2f}")

# :, = just commas
print(f"Large Number: €{large_number:,}")

# .1% = percentage with 1 decimal
print(f"Margin: {margin:.1%}")
```

## 4. Booleans and Comparisons

Booleans are True or False values. Comparison operators create booleans.

```python
# Demo: Comparison operators
a = 10
b = 5

print(f"a == b: {a == b}")  # Equal to
print(f"a != b: {a != b}")  # Not equal to
print(f"a > b: {a > b}")    # Greater than
print(f"a < b: {a < b}")    # Less than
print(f"a >= 10: {a >= 10}") # Greater than or equal
print(f"b <= 5: {b <= 5}")   # Less than or equal
```

## 5. Type Checking and Conversion

Use `type()` to check types and `int()`, `float()`, `str()` to convert.

```python
# Demo: Type conversion
# When reading from CSV files, all values come as strings!

units_str = "25"
price_str = "199.00"

# Convert to numbers
units = int(units_str)
price = float(price_str)

# Now we can do math
total = units * price

print(f"Units: {units} (type: {type(units).__name__})")
print(f"Price: {price} (type: {type(price).__name__})")
print(f"Total: €{total:,.2f}")
```
