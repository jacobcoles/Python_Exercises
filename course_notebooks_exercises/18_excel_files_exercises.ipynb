## 1. Setup - Create Sample Excel File

```python
# Create sample Excel file to work with
import pandas as pd

# Create sample data
q1_sales = pd.DataFrame({
    "Product": ["Suite A", "Suite B", "Analytics"],
    "Revenue": [45000, 32000, 89000],
    "Units": [300, 400, 150]
})

q2_sales = pd.DataFrame({
    "Product": ["Suite A", "Suite B", "Analytics"],
    "Revenue": [52000, 38000, 95000],
    "Units": [350, 450, 165]
})

# Write to Excel with multiple sheets
with pd.ExcelWriter("sample_sales.xlsx") as writer:
    q1_sales.to_excel(writer, sheet_name="Q1", index=False)
    q2_sales.to_excel(writer, sheet_name="Q2", index=False)

print("Sample Excel file created!")
```

## 2. Reading Excel Files

```python
# Demo: Read Excel file
import pandas as pd

# Read first sheet (default)
df = pd.read_excel("sample_sales.xlsx")
print("First sheet (default):")
print(df)

# Read specific sheet by name
q2 = pd.read_excel("sample_sales.xlsx", sheet_name="Q2")
print("\nQ2 sheet:")
print(q2)
```
