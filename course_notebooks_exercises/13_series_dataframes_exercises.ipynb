## 2. Selection with loc (Label-Based)

```python
# Demo: loc selection
print("Row 0:")
print(products.loc[0])

print("\nRows 0-2, ProductName and Price:")
print(products.loc[0:2, ["ProductName", "Price"]])
```

---
## 3. Selection with iloc (Position-Based)

```python
# Demo: iloc selection
print("First row (iloc[0]):")
print(products.iloc[0])

print("\nLast row (iloc[-1]):")
print(products.iloc[-1])

print("\nFirst 3 rows, first 2 cols:")
print(products.iloc[0:3, 0:2])
```

## 4. Boolean Selection (Filtering)

```python
# Demo: Boolean filtering
# Price > 200
expensive = products[products["Price"] > 200]
print("Price > 200:")
print(expensive)

# Combining conditions with &
premium_expensive = products[
    (products["Category"] == "Premium") & 
    (products["Price"] > 200)
]
print("\nPremium AND Price > 200:")
print(premium_expensive)
```

## 5. Setting Values

```python
# Demo: Setting values with loc
df = products.copy()

# Add new column
df["Discount"] = 0.10

# Update based on condition
df.loc[df["Category"] == "Premium", "Discount"] = 0.15

print(df)
```
