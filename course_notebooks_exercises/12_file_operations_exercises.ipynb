## 1. Setup: Create Sample File

```python
# Create a sample CSV file to work with
import csv

sample_data = [
    {"Product": "Suite A", "Revenue": "1500", "Units": "10"},
    {"Product": "Suite B", "Revenue": "990", "Units": "15"},
    {"Product": "Analytics", "Revenue": "3990", "Units": "5"},
]

with open("sales.csv", "w", newline="") as file:
    writer = csv.DictWriter(file, fieldnames=["Product", "Revenue", "Units"])
    writer.writeheader()
    writer.writerows(sample_data)

print("Sample CSV created!")
```

## 2. Reading CSV Files

```python
# Demo: Reading CSV with csv module
import csv

with open("sales.csv", "r") as file:
    reader = csv.DictReader(file)
    for row in reader:
        print(f"{row['Product']}: €{row['Revenue']}")
```

## 3. Writing CSV Files

```python
# Demo: Writing CSV
import csv

data = [
    {"Product": "New A", "Price": 299},
    {"Product": "New B", "Price": 199}
]

with open("output.csv", "w", newline="") as file:
    writer = csv.DictWriter(file, fieldnames=["Product", "Price"])
    writer.writeheader()
    writer.writerows(data)

print("File written!")
```
