## 3. Common AI Use Cases

```python
# Use Case 1: Ask AI to Explain Code
# This is one of the BEST uses of AI

# You see this code and don't understand it:
df_sorted = df.sort_values(['Region', 'Revenue'], ascending=[True, False])

# Good question to AI:
# "Can you explain this line step by step?
#  df.sort_values(['Region', 'Revenue'], ascending=[True, False])
#  What does ascending=[True, False] mean?"

print("Original:")
print(df)
print("\nSorted:")
print(df_sorted)
```

```python
# Use Case 2: Ask AI to Debug Errors
# Include the FULL error message and context

# Example: You get KeyError: 'Sales'

df = pd.DataFrame({
    'Region': ['EMEA', 'AMER'],
    'Revenue': [15000, 12000]  # Note: column is 'Revenue' not 'Sales'
})

# This will cause an error:
# total = df['Sales'].sum()

# Good question to AI:
# "I'm getting KeyError: 'Sales' on this line:
#  total = df['Sales'].sum()
#  My DataFrame columns are: ['Region', 'Revenue']
#  What's wrong?"

# Check available columns:
print("Available columns:", df.columns.tolist())
```

## 4. AI Limitations

```python
# AI Limitation 1: Hallucination - Making up methods that don't exist

# AI might suggest:
# df.smart_analyze()  # This doesn't exist!

# Always verify methods exist in official documentation
# pandas.pydata.org/docs

print("Tip: If you're unsure if a method exists, try help(df.method_name)")
print("Or check: dir(df) to see all available methods")
```

```python
# AI Limitation 2: Outdated Syntax

# Old way (AI might suggest, now deprecated):
# df.append(new_row)  # Deprecated in newer Pandas!

# Better way:
new_row = pd.DataFrame({'Region': ['APAC'], 'Revenue': [9500]})
df_updated = pd.concat([df, new_row], ignore_index=True)

print("Updated DataFrame:")
print(df_updated)
```
