## 1. String Cleaning Methods

```python
# Demo: Essential cleaning methods

# strip() - remove leading/trailing whitespace
product = "  Suite A  "
clean = product.strip()
print(f"'{product}' -> '{clean}'")

# upper() and lower() - change case
region = "emea"
print(f"Upper: {region.upper()}")

# replace() - substitute characters
code = "SUITE-A-2024"
clean_code = code.replace("-", "_")
print(f"Cleaned: {clean_code}")
```

## 2. Checking String Properties

```python
# Demo: Checking strings
product_code = "SUITE-A-2024"

# startswith() and endswith()
print(f"Starts with 'SUITE': {product_code.startswith('SUITE')}")
print(f"Ends with '2024': {product_code.endswith('2024')}")

# isdigit() - check if only digits
order_id = "12345"
print(f"'{order_id}' is numeric: {order_id.isdigit()}")

# 'in' operator - check substring
email = "john@acme.com"
print(f"Contains '@': {'@' in email}")
```

## 3. Splitting and Joining

```python
# Demo: Split and join

# split() - break string into list
csv_line = "Suite A,199,150"
fields = csv_line.split(",")
print(f"Fields: {fields}")

# join() - combine list into string
regions = ["EMEA", "AMER", "APAC"]
csv = ",".join(regions)
print(f"CSV: {csv}")
```
