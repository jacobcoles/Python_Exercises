## 4. Common Pitfalls to Avoid

```python
# Pitfall 1: Modifying Original Data
import pandas as pd

# BAD: Original data lost!
# df = pd.read_csv('sales.csv')
# df = df[df['revenue'] > 1000]  # Original gone!

# GOOD: Keep original, create filtered version
df_raw = pd.DataFrame({
    'Product': ['A', 'B', 'C'],
    'Revenue': [500, 1500, 2000]
})

df_filtered = df_raw[df_raw['Revenue'] > 1000]  # Original preserved

print("Original:")
print(df_raw)
print("\nFiltered:")
print(df_filtered)
```

```python
# Pitfall 2: Hardcoding Values

# BAD: What is 0.80?
revenue_after_tax = 15000 * 0.80

# GOOD: Use named constants
TAX_RATE = 0.20
revenue_after_tax = 15000 * (1 - TAX_RATE)

print(f"Revenue after {TAX_RATE*100}% tax: ${revenue_after_tax:,}")
```

```python
# Pitfall 3: Not Handling Errors

# BAD: Crashes if revenue is 0!
# margin = profit / revenue

# GOOD: Handle edge case
profit = 500
revenue = 0

if revenue == 0:
    margin = 0
else:
    margin = profit / revenue

print(f"Margin: {margin}")
```
