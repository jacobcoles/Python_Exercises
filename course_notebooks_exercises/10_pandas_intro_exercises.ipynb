## 2. Loading CSV Files

```python
# Demo: Create and read CSV
import pandas as pd

# Create sample CSV
sample = pd.DataFrame({
    "Date": ["2024-01-15", "2024-01-16", "2024-01-17"],
    "Product": ["Suite A", "Suite B", "Suite A"],
    "Revenue": [1500, 990, 1500]
})
sample.to_csv("sample_sales.csv", index=False)

# Read CSV back
df = pd.read_csv("sample_sales.csv")
print(df)
```
