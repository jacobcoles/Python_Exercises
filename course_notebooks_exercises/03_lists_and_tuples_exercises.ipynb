## 1. Creating and Analyzing Lists

```python
# Demo: Why Lists Are Powerful
revenues = [15000, 18000, 16500, 19000, 17500, 21000]

# Built-in functions work on entire lists
total = sum(revenues)
average = sum(revenues) / len(revenues)
maximum = max(revenues)
minimum = min(revenues)

print(f"Total revenue: €{total:,}")
print(f"Average: €{average:,.2f}")
print(f"Best month: €{maximum:,}")
print(f"Worst month: €{minimum:,}")
```

## 2. Indexing - Accessing Individual Elements

**Important**: Python counts from 0, not 1!

```python
# Demo: Positive and Negative Indexing
products = ["Suite A", "Suite B", "Analytics", "Basic", "Premium"]
#            0          1           2          3         4
#           -5         -4          -3         -2        -1

print(f"First (index 0): {products[0]}")
print(f"Third (index 2): {products[2]}")
print(f"Last (index -1): {products[-1]}")
print(f"Second to last (index -2): {products[-2]}")
```

## 3. Slicing - Extracting Portions of Lists

Syntax: `list[start:end]` - from start up to (but NOT including) end

```python
# Demo: Slicing
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]

q1 = months[0:3]   # First 3 (indices 0, 1, 2)
q2 = months[3:6]   # Next 3 (indices 3, 4, 5)
first_half = months[:6]   # First 6 (omit start = from beginning)
last_3 = months[-3:]      # Last 3 (omit end = to the end)

print(f"Q1: {q1}")
print(f"Q2: {q2}")
print(f"First half: {first_half}")
print(f"Last 3: {last_3}")
```

## 4. List Methods - Modifying Lists

```python
# Demo: Adding and Removing Items
revenues = [15000, 18000, 16500]
print(f"Original: {revenues}")

# append() - add ONE item to end
revenues.append(19000)
print(f"After append: {revenues}")

# extend() - add MULTIPLE items
revenues.extend([17500, 21000])
print(f"After extend: {revenues}")

# remove() - remove by VALUE
revenues.remove(16500)
print(f"After remove: {revenues}")

# sort() - sort in place
revenues.sort()
print(f"After sort: {revenues}")
```

## 5. Checking Membership with `in`

```python
# Demo: The 'in' operator
valid_regions = ["AMER", "EMEA", "APAC", "NORD"]

user_region = "EMEA"
print(f"Is {user_region} valid? {user_region in valid_regions}")

user_region = "LATAM"
print(f"Is {user_region} valid? {user_region in valid_regions}")
```

## 6. Tuples - Immutable Lists

```python
# Demo: Tuples
# Use parentheses () instead of brackets []
# Tuples cannot be changed after creation

product = ("PROD-001", "Suite A", 199, "Premium")

# Access items the same way as lists
print(f"ID: {product[0]}")
print(f"Name: {product[1]}")
print(f"Price: €{product[2]}")

# Tuple unpacking
prod_id, name, price, tier = product
print(f"\nUnpacked: {name} costs €{price}")
```
