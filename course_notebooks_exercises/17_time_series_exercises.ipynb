## 1. Converting to Datetime

```python
# Demo: Convert strings to datetime
import pandas as pd

df = pd.DataFrame({
    'Date': ['2024-01-15', '2024-01-16', '2024-01-17'],
    'Revenue': [1500, 2000, 1800]
})

print("Before (object type):")
print(df.dtypes)

# Convert to datetime
df['Date'] = pd.to_datetime(df['Date'])

print("\nAfter (datetime64):")
print(df.dtypes)
```

## 2. Extracting Date Components

```python
# Demo: Extract year, month, quarter
df = pd.DataFrame({
    'Date': pd.to_datetime(['2024-01-15', '2024-04-20', '2024-07-10']),
    'Revenue': [15000, 18000, 21000]
})

df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Quarter'] = df['Date'].dt.quarter

print(df)
```

## 3. Filtering by Date Ranges

```python
# Demo: Filter by date range
dates = pd.date_range('2024-01-01', '2024-12-31', freq='D')
df = pd.DataFrame({
    'Date': dates,
    'Revenue': range(1000, 1000 + len(dates))
})

# Filter Q1
q1 = df[df['Date'].between('2024-01-01', '2024-03-31')]
print(f"Q1 records: {len(q1)}")
print(f"Q1 total: €{q1['Revenue'].sum():,}")
```

## 4. Rolling Calculations

```python
# Demo: Rolling average
import numpy as np

dates = pd.date_range('2024-01-01', periods=14, freq='D')
df = pd.DataFrame({
    'Date': dates,
    'Sales': [2000 + np.random.randint(-300, 500) for _ in range(14)]
})

# 7-day rolling average
df['Rolling_7Day'] = df['Sales'].rolling(window=7).mean()

print(df)
```

## 5. Time-Based Aggregations

```python
# Demo: Resample to monthly
dates = pd.date_range('2024-01-01', '2024-06-30', freq='D')
daily = pd.DataFrame({
    'Date': dates,
    'Revenue': [1500 + i*2 for i in range(len(dates))]
}).set_index('Date')

# Resample to monthly totals
monthly = daily['Revenue'].resample('M').sum()
print("Monthly totals:")
print(monthly)
```
