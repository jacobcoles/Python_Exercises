## 2. Essential Keyboard Shortcuts

### Running Cells:
- **Shift + Enter** - Run cell and move to next cell
- **Ctrl + Enter** - Run cell and stay on same cell

### Adding/Deleting Cells (press Esc first):
- **A** - Insert cell Above
- **B** - Insert cell Below
- **DD** - Delete cell (press D twice)

### Cell Types (press Esc first):
- **Y** - Change to code cell
- **M** - Change to markdown cell

## 4. Comments in Code

### Why Use Comments?
- Explain what your code does
- Leave notes for yourself and others

```python
# Demo: Single-line comments
# This is a comment - Python ignores it

revenue = 50000  # You can also add comments at end of lines

# Calculate 20% VAT
vat = revenue * 0.20

print(f"Revenue: €{revenue:,}")
print(f"VAT: €{vat:,}")
```

### Exercise 4: Document Your Code

The code below calculates a discounted price but has no comments.

```python
# Exercise 4: Add Comments to Code
# Add comments above or beside each line to explain what it does

list_price = 199.00
units = 50
discount = 0.15
subtotal = list_price * units
discount_amount = subtotal * discount
final_price = subtotal - discount_amount

print(f"Final Price: €{final_price:,.2f}")
```
