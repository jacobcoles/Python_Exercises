## 1. Basic Try/Except

```python
# Demo: Safe data conversion
values = ["100", "200", "invalid", "300"]

total = 0
errors = 0

for value in values:
    try:
        number = int(value)
        total += number
    except ValueError:
        print(f"Warning: Cannot convert '{value}'")
        errors += 1

print(f"\nTotal: {total}")
print(f"Errors: {errors}")
```

## 2. Handling Multiple Error Types

```python
# Demo: Multiple except blocks
def safe_calculate(revenue_str, cost_str):
    try:
        revenue = float(revenue_str)
        cost = float(cost_str)
        margin = ((revenue - cost) / revenue) * 100
        return f"Margin: {margin:.1f}%"
    except ValueError:
        return "Error: Invalid number format"
    except ZeroDivisionError:
        return "Error: Revenue cannot be zero"

print(safe_calculate("1000", "600"))     # Normal
print(safe_calculate("invalid", "600"))  # ValueError
print(safe_calculate("0", "600"))        # ZeroDivisionError
```

## 3. Try/Except/Else/Finally

```python
# Demo: Complete error handling
def process_transaction(amount_str):
    try:
        amount = float(amount_str)
        if amount <= 0:
            raise ValueError("Amount must be positive")
    except ValueError as e:
        print(f"Error: {e}")
        return None
    else:
        # Only runs if try succeeded
        print(f"Transaction processed: €{amount:,.2f}")
        return amount
    finally:
        # Always runs (for cleanup)
        print("Transaction attempt logged\n")

process_transaction("1500")
process_transaction("invalid")
process_transaction("-100")
```

## 4. Common Python Errors

```python
# Demo: Common Python error types

# 1. ValueError - wrong value format
try:
    number = int("not a number")
except ValueError:
    print("ValueError: Cannot convert to integer")

# 2. ZeroDivisionError
try:
    result = 10 / 0
except ZeroDivisionError:
    print("ZeroDivisionError: Cannot divide by zero")

# 3. KeyError - missing dictionary key
try:
    data = {"name": "Product A"}
    price = data["price"]
except KeyError:
    print("KeyError: Key 'price' not found")

# 4. IndexError - list index out of range
try:
    items = [1, 2, 3]
    value = items[10]
except IndexError:
    print("IndexError: List index out of range")
```
