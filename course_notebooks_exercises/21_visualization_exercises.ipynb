```python
# Setup: Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

%matplotlib inline
plt.style.use('seaborn-v0_8-darkgrid')
plt.rcParams['figure.figsize'] = (10, 6)
```

## 1. Line Charts - Showing Trends

```python
# Demo: Basic Line Chart
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
revenue = [15000, 18000, 16500, 19000, 17500, 21000]

df = pd.DataFrame({'Month': months, 'Revenue': revenue})

# Create line chart
df.plot(x='Month', y='Revenue', kind='line', marker='o')

plt.title('Monthly Revenue Trend')
plt.ylabel('Revenue')
plt.show()
```

## 2. Bar Charts - Comparing Categories

```python
# Demo: Vertical Bar Chart
regions = ['EMEA', 'AMER', 'APAC']
sales = [350000, 280000, 245000]

df = pd.DataFrame({'Region': regions, 'Sales': sales})

# Create vertical bar chart
df.plot(x='Region', y='Sales', kind='bar', color='steelblue')

plt.title('Sales by Region')
plt.ylabel('Total Sales')
plt.xticks(rotation=0)
plt.show()
```

## 3. Scatter Plots - Showing Relationships

```python
# Demo: Scatter Plot
# Scatter plots show relationship between two numeric variables

df = pd.DataFrame({
    'Marketing': [5000, 8000, 6500, 12000, 9500, 15000, 11000, 7500],
    'Revenue': [45000, 68000, 52000, 95000, 78000, 118000, 88000, 62000]
})

# Create scatter plot
df.plot(kind='scatter', x='Marketing', y='Revenue', s=100, alpha=0.6)

plt.title('Marketing Spend vs Revenue')
plt.xlabel('Marketing Spend')
plt.ylabel('Revenue')
plt.grid(True, alpha=0.3)
plt.show()
```

## 4. Histograms and Box Plots

```python
# Demo: Histogram - Distribution of Values
np.random.seed(42)
transactions = np.random.normal(2500, 800, 200)

df = pd.DataFrame({'Amount': transactions})

# Create histogram
df['Amount'].plot(kind='hist', bins=20, color='skyblue', edgecolor='black')

plt.title('Transaction Amount Distribution')
plt.xlabel('Transaction Amount')
plt.ylabel('Frequency')
plt.axvline(df['Amount'].mean(), color='red', linestyle='--', label='Mean')
plt.legend()
plt.show()
```

```python
# Demo: Box Plot - Statistical Summary
np.random.seed(42)
df = pd.DataFrame({
    'EMEA': np.random.normal(3000, 500, 50),
    'AMER': np.random.normal(2500, 600, 50),
    'APAC': np.random.normal(2200, 400, 50)
})

# Create box plot
df.plot(kind='box')

plt.title('Revenue Distribution by Region')
plt.ylabel('Transaction Amount')
plt.show()
```

## 5. Multi-Chart Dashboards

```python
# Demo: 2x2 Dashboard
# plt.subplots() creates a grid of charts

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
revenue = [15000, 18000, 16500, 19000, 17500, 21000]
regions = ['EMEA', 'AMER', 'APAC']
regional_sales = [350000, 280000, 245000]

# Create 2x2 grid
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Chart 1: Line chart
axes[0, 0].plot(months, revenue, marker='o', color='green')
axes[0, 0].set_title('Monthly Trend')
axes[0, 0].set_ylabel('Revenue')

# Chart 2: Bar chart
axes[0, 1].bar(regions, regional_sales, color=['steelblue', 'coral', 'lightgreen'])
axes[0, 1].set_title('Sales by Region')

# Chart 3: Histogram
axes[1, 0].hist(np.random.normal(18000, 2000, 100), bins=15, color='skyblue', edgecolor='black')
axes[1, 0].set_title('Revenue Distribution')

# Chart 4: Scatter
axes[1, 1].scatter([99, 199, 299, 499], [850, 520, 280, 120], s=200, alpha=0.6, color='purple')
axes[1, 1].set_title('Price vs Units')

plt.suptitle('Sales Dashboard', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()
```
