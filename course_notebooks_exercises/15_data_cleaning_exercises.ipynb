## 1. Handling Missing Data

```python
# Demo: Detecting and handling missing values
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Product": ["A", "B", "C", "D"],
    "Price": [199, np.nan, 299, 99]
})

# Detect missing values
print("Missing values:")
print(df.isna())

# Fill missing values
df["Price"] = df["Price"].fillna(0)
print("\nAfter filling with 0:")
print(df)
```

## 2. Missing Value Strategies

```python
# Demo: Different strategies
df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [10000, np.nan, 12000, np.nan, np.nan, 15000]
})

print("Original:")
print(df)

# Strategy 1: Fill with constant
df1 = df.copy()
df1["Sales"] = df1["Sales"].fillna(0)
print("\nFilled with 0:")
print(df1)

# Strategy 2: Fill with mean
df2 = df.copy()
mean_value = df["Sales"].mean()
df2["Sales"] = df2["Sales"].fillna(mean_value)
print(f"\nFilled with mean ({mean_value:,.0f}):")
print(df2)

# Strategy 3: Forward fill (ffill)
df3 = df.copy()
df3["Sales"] = df3["Sales"].ffill()
print("\nForward filled:")
print(df3)
```

## 3. Removing Duplicates

```python
# Demo: Remove duplicates
df = pd.DataFrame({
    "Product": ["A", "B", "A", "C"],
    "Price": [199, 99, 199, 299]
})

print("Original (with duplicate):")
print(df)

# Remove duplicate rows
df_clean = df.drop_duplicates()
print("\nAfter removing duplicates:")
print(df_clean)
```
