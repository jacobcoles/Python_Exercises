## 1. Basic For Loops

```python
# Demo: Calculate total using a loop
revenues = [3082.59, 556.34, 2422.68, 4155.00, 2486.50]

# Initialize accumulator BEFORE the loop
total_revenue = 0

# Loop through each value
for revenue in revenues:
    total_revenue += revenue  # Add to running total

print(f"Total Revenue: €{total_revenue:,.2f}")
```

## 3. Enumerate() - Track Positions

```python
# Demo: enumerate() gives index AND value
products = ["Basic", "Plus", "Pro", "Enterprise"]

for index, product in enumerate(products):
    print(f"{index+1}. {product}")

# Or use start=1
print("\nWith start=1:")
for number, product in enumerate(products, start=1):
    print(f"{number}. {product}")
```

## 4. Building Lists with Loops

```python
# Demo: Filter items with a condition
all_amounts = [3082.59, 556.34, 2422.68, 4155.00, 2486.50, 15000.00]

large_transactions = []

for amount in all_amounts:
    if amount > 3000:
        large_transactions.append(amount)

print(f"All amounts: {all_amounts}")
print(f"Large transactions: {large_transactions}")
```

## 5. While Loops

```python
# Demo: Accumulate until target reached
daily_sales = [1200, 1450, 980, 1680, 1320, 1890, 2100, 1560]
target = 5000
total = 0
day = 0

while total < target and day < len(daily_sales):
    total += daily_sales[day]
    day += 1
    print(f"Day {day}: €{daily_sales[day-1]:,}, Total: €{total:,}")

if total >= target:
    print(f"\nTarget reached on Day {day}!")
else:
    print(f"\nTarget not reached after {day} days")
```

## 6. Break and Continue

```python
# Demo: break - exit loop when found
transactions = [1500, 800, 2200, 5500, 1200, 3400]

print("Searching for first transaction over €5000...")
for amount in transactions:
    print(f"Checking €{amount:,}")
    if amount > 5000:
        print(f"Found it! €{amount:,}")
        break  # Exit immediately
```

```python
# Demo: continue - skip invalid items
revenues = [3500, -100, 2800, 0, 4200, -50, 1900]

print("Processing only positive values:")
total = 0
for revenue in revenues:
    if revenue <= 0:
        print(f"Skipping: €{revenue}")
        continue  # Skip to next iteration
    total += revenue
    print(f"Added €{revenue:,}")

print(f"Total: €{total:,}")
```

## 7. List Comprehensions (Optional)

A concise way to create lists.

```python
# Demo: List comprehensions
prices = [49, 99, 199, 499]

# Transform: apply 10% increase
increased = [price * 1.10 for price in prices]
print(f"Increased: {increased}")

# Filter: keep only > 100
large = [p for p in prices if p > 100]
print(f"Large prices: {large}")
```
